## 1. Parameters: giving a function input

Yesterday's header could only print *one* café name. A **parameter** is a variable in the `def` line that
receives a value each time the function is called.

```python
def print_receipt_header(shop_name):
    """Print a receipt header for the shop called shop_name."""
    print("=" * 30)
    print(f"  {shop_name}")
    print("=" * 30)
```

```python
print_receipt_header("CORNER CAFE")
print_receipt_header("BOOK NOOK")
```

| Word | Where | Here |
|------|-------|------|
| **parameter** | the name in the `def` line (the blank on a form) | `shop_name` |
| **argument** | the value in the call (what you write in the blank) | `"CORNER CAFE"` |

🤔 **Predict:**

```python
def greet(name):
    """Print a greeting for name."""
    print(f"Hello, {name}! Welcome back.")


greet("Sara")
greet("Diego")
```

## 2. Several parameters: positional and keyword arguments

```python
def print_order(item, quantity, size):
    """Print one line of a drinks order."""
    print(f"{quantity} x {size} {item}")


print_order("latte", 2, "large")
```

**Positional arguments** are matched by **position**: 1st argument → 1st parameter, and so on.

🤔 **Predict:** what happens if we mix up the order?

```python
print_order(2, "latte", "large")
```

No error, just nonsense: Python cannot know what we *meant*. **Keyword arguments** name the parameter, so the
order no longer matters and the call explains itself:

```python
print_order(size="small", item="tea", quantity=3)
print_order("mocha", size="medium", quantity=1)  # mixing: positional ones first
```

## 3. `return`: giving a result back

`return value` ends the function and **hands the value back** to the line that called it.
The call is then replaced by that value, so we can store it, print it or calculate with it.

```python
def km_to_miles(km):
    """Return the distance km converted to miles."""
    return km * 0.621371


distance = km_to_miles(10)
print(distance)
```

```python
print(f"A 5 km run is {km_to_miles(5):.2f} miles.")
total = km_to_miles(3) + km_to_miles(7)
print(f"3 km + 7 km = {total:.2f} miles")
```

🤔 **Predict** the value before running:

```python
def celsius_to_fahrenheit(celsius):
    """Return the temperature celsius converted to Fahrenheit."""
    return celsius * 9 / 5 + 32


celsius_to_fahrenheit(100)
```

## 4. `print` vs `return`: the most important idea today

Two functions that *look* the same when you call them:

```python
def add_service_print(bill):
    """Print the bill plus a 10% service charge."""
    print(bill + bill * 0.1)


def add_service(bill):
    """Return the bill plus a 10% service charge."""
    return bill + bill * 0.1
```

```python
add_service_print(50)
```

🤔 **Predict:** we store the "result" of the printing version. What is inside `total`?

```python
total = add_service_print(50)
print("total is:", total)
```

The function **showed** `55.0` on the screen, but gave **nothing back**. "Nothing" in Python is `None`.
Now try to split that bill between two people:

```python
each = total / 2
```

With the returning version, the value comes back and we can keep calculating:

```python
total = add_service(50)
each = total / 2
print("Each person pays", each)
```

> **Analogy:** `print` is a waiter **telling** you the price: you heard it, but you cannot do maths with it later.
> `return` **hands you the receipt**: you can keep it, split it, add it to another one.

**Rule of thumb:** a function that *calculates* should `return` the result. Let the code that called it decide
whether to print it.

## 5. `None`: the value that means "nothing"

Every function returns something. If it has no `return`, it returns `None`.

```python
def print_divider():
    """Print a line of 30 dashes."""
    print("-" * 30)


result = print_divider()
print(result)
print(type(result))
```

## 6. `return` ends the function immediately

Once a `return` runs, the function stops. Nothing after it inside the function runs.

```python
def letter_grade(score):
    """Return the letter grade for a score out of 100."""
    if score >= 80:
        return "A"
    if score >= 70:
        return "B"
    if score >= 60:
        return "C"
    if score >= 50:
        return "D"
    return "F"


print(letter_grade(91), letter_grade(74), letter_grade(50), letter_grade(12))
```

We do not need `elif` here: as soon as one `return` runs, the function is finished, so later `if`s are never
reached. The last `return "F"` only runs if no earlier one did.

A comparison is already `True` or `False`, so it can be returned directly:

```python
def is_passing(score):
    """Return True if score is 50 or more."""
    return score >= 50


print(is_passing(72), is_passing(35))
```

## 7. Returning two values (a tuple)

Separate the values with a comma after `return`. Python packs them into a **tuple** (Day 07), and we can
**unpack** it into two variables.

```python
def apply_discount(price, percent):
    """Return the price after a percent discount, and the amount saved."""
    saving = price * percent / 100
    return price - saving, saving


result = apply_discount(80, 25)
print(result)
```

```python
new_price, saved = apply_discount(80, 25)
print(f"Pay {new_price:.2f}, you save {saved:.2f}")
```

## 8. Common mistakes

**Mistake 1: calling a returning function and ignoring the result.** In a `.py` file (or anywhere except the last
line of a notebook cell) the value is simply thrown away:

```python
km_to_miles(10)
print("done")
```

Only `done` appears. Store the result (`miles = km_to_miles(10)`) or print it.

The next cells fail **on purpose**:

```python
# Mistake 2: forgetting an argument
km_to_miles()
```

```python
# Mistake 3: too many arguments
km_to_miles(5, 10)
```

```python
# Mistake 4: a positional argument after a keyword argument
print_order(item="tea", 2, "small")
```

(Mistake 5, using the result of a function that only prints, is in section 4.)

## 9. Real-world example: a grade report

`letter_grade` returns a value, so we can use it inside an f-string, in a loop, anywhere.

```python
marks = {"Asha": 91, "Bilal": 67, "Mei": 74, "Omar": 48}

for name, score in marks.items():
    print(f"{name:<6} {score:>3}  {letter_grade(score)}  passed: {is_passing(score)}")
```

## Recap

1. **Parameters** are the names in the `def` line; **arguments** are the values in the call.
2. Positional arguments match by order; keyword arguments (`size="small"`) match by name.
3. `return value` ends the function and hands the value back to the caller.
4. `print` only *shows* a value. A function without `return` gives back `None`.
5. `return a, b` gives back a tuple; unpack it with `x, y = f(...)`.
