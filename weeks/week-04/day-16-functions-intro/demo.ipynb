## 1. The duplication problem

A café prints a receipt for every customer. Every receipt starts with the same header.

````python
print("=" * 30)
print("        CORNER CAFE")
print("   12 Garden Road, Lahore")
print("=" * 30)
print("Customer: Asha    Latte  4.50")
print()
print("=" * 30)
print("        CORNER CAFE")
print("   12 Garden Road, Lahore")
print("=" * 30)
print("Customer: Bilal   Tea    2.00")
````

🤔 The café changes its name to *Corner Cafe & Books*. **How many lines do we have to edit?**
What happens if we forget one of them?

With 2 customers it is annoying. With 50 receipts (or a 200-line Quiz Game) it is a bug waiting to happen.
We want to write the header **once**, give it a **name**, and use that name whenever we need it.

## 2. Defining a function

```
def name():          <- "def" + a name + () + a colon
    line 1           <- the body: indented 4 spaces
    line 2
```

````python
def print_receipt_header():
    print("=" * 30)
    print("        CORNER CAFE")
    print("   12 Garden Road, Lahore")
    print("=" * 30)
````

🤔 **Predict:** the cell above ran. Why did nothing appear?

`def` only **teaches** Python the recipe and stores it under the name `print_receipt_header`.
Nothing inside it runs yet. To run it, we **call** it: its name followed by `()`.

````python
print_receipt_header()
````

Now the receipts. Each call runs the whole body again.

````python
print_receipt_header()
print("Customer: Asha    Latte  4.50")
print()
print_receipt_header()
print("Customer: Bilal   Tea    2.00")
````

To rename the café we now change **one line** inside the function, re-run the `def` cell, and every receipt
uses the new name.

## 3. Calling many times: the order things run in

````python
def say_hello():
    print("Hello!")
````

🤔 **Predict** the exact order of the four lines before running:

````python
print("1. before")
say_hello()
print("2. after")
say_hello()
````

When Python reaches a call it **jumps into the function**, runs the body, then **comes back** to the line
after the call. We can also call a function inside a loop:

````python
def print_divider():
    print("-" * 30)


for table in range(1, 4):
    print("Table", table)
    print_divider()
````

## 4. Docstrings: saying what a function does

The first line of the body can be a string in triple quotes: a **docstring**. It is not printed. It
documents *what the function does* for anyone who uses it. In VS Code, hover over a function call to see it.

````python
def print_divider():
    """Print a line of 30 dashes."""
    print("-" * 30)


print_divider()
````

Good habits:

| Habit | Example |
|-------|---------|
| Name says what it **does**, usually a verb, `snake_case` | `print_divider`, `show_menu`, `draw_box` |
| One-sentence docstring that starts with a verb | `"""Print a line of 30 dashes."""` |
| `#` comments explain *how* or *why* inside the body | `# 30 fits the receipt printer width` |

Never name a function `print`, `input` or `type`: you would hide Python's own.

## 5. A function body can contain anything: loops, ifs, variables

````python
def show_menu():
    """Print the café menu as a numbered list."""
    items = ["Tea", "Coffee", "Hot chocolate", "Lemonade"]
    print("--- MENU ---")
    for number, item in enumerate(items, start=1):
        print(f"{number}. {item}")


show_menu()
````

🤔 **Predict:** how many lines does the next cell print?

````python
for customer in ["Mei", "Omar"]:
    print("Hi,", customer)
    show_menu()
````

Two greetings + two menus of 5 lines = **12 lines**. Five lines of code inside `show_menu` did the work of ten.

## 6. Common mistakes

**Mistake 1: forgetting the parentheses.** No error, but the function does **not** run.
The notebook just shows what the name *is*:

````python
print_divider
````

`print_divider` is the function itself; `print_divider()` **runs** it. In a `.py` file this line silently does nothing.

The next cells fail **on purpose**. Read the last line of each error.

````python
# Mistake 2: calling a function before it has been defined
print_footer()
````

````python
def print_footer():
    """Print the last line of a receipt."""
    print("Thank you, come again!")


print_footer()
````

Python reads top to bottom. In a `.py` file, put all the `def`s **above** the code that calls them.
In a notebook, run the `def` cell first.

````python
# Mistake 3: missing colon after the parentheses
def print_title()
    print("Daily Specials")
````

````python
# Mistake 4: the body is not indented
def print_title():
print("Daily Specials")
````

## 7. Real-world example: the café screen reads like a story

Once the details live inside functions, the main part of the program reads like a list of steps.

````python
def show_welcome():
    """Greet the customer."""
    print("Welcome to Corner Cafe!")


def show_goodbye():
    """Thank the customer."""
    print("Thank you for visiting. See you soon!")
````

````python
choices = ["2", "4", "0"]  # pretend the customer typed these

show_welcome()
show_menu()
for choice in choices:
    if choice == "0":
        show_goodbye()
        break
    print("You ordered item number", choice)
````

Notice we can only print the item *number*: the list of item names lives **inside** `show_menu`, and the rest of
the program cannot see it. Getting information **into** and **out of** a function is tomorrow's topic.

## Recap

1. A function is a **named block of code**. Write it once, call it as often as you like.
2. `def name():` + indented body **defines** it (nothing runs). `name()` **calls** it.
3. A call jumps into the function, runs the body, then comes back to the next line.
4. A docstring (`"""..."""`, first line of the body) says what the function does.
5. Forgetting `()` gives no error but runs nothing; calling before defining gives a `NameError`.
