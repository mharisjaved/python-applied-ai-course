## Exercise 2: Write the checks

`count_vowels` below works. Write **at least four** checks for it:

1. a normal word,
2. a word with capital letters,
3. a word with no vowels,
4. the empty string `""`.

Every check should print `True`.

````python
def count_vowels(text):
    """Return how many vowels (a, e, i, o, u, in any case) text contains."""
    count = 0
    for char in text.lower():
        if char in "aeiou":
            count += 1
    return count


# your checks here
````
