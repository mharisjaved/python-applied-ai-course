## Exercise 4: Let the tests find the bug

`count_vowels` **has a bug**, but this check passes:

```python
assert count_vowels("banana") == 3
```

Write test functions with inputs that show the bug (think about what the docstring promises), run them, read the
failure, then fix `count_vowels` and run **all** your tests again.

````python
def count_vowels(text):
    """Return how many vowels (a, e, i, o, u) are in text, in upper or lower case."""
    count = 0
    for letter in text:
        if letter in "aeiou":
            count += 1
    return count


def test_count_vowels_normal():
    assert count_vowels("banana") == 3


test_count_vowels_normal()
print("normal test passed")
````
