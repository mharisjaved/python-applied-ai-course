## 1. Why modules?

So far each program lived in **one** notebook or file. That stops working when:

- the file grows to hundreds of lines and you cannot find anything;
- you copy-paste the same function into five notebooks, then fix a bug in only one of them.

A **module** is simply a `.py` file full of functions and variables. `import` lets another file use them.
Python ships with hundreds of ready-made modules (the **standard library**). You already used one on Day 14: `random`.

## 2. `import module`

`import math` loads the module once. Then you use its contents with **module name + dot**: `math.sqrt`.

```python
import math

print(math.sqrt(81))
print(math.pi)
```

🤔 **Predict:** `ceil` rounds **up** to a whole number, `floor` rounds **down**. What do these show?

```python
print(math.ceil(4.2))
print(math.floor(4.8))
```

A real use: 47 students go on a trip and each minibus has 12 seats. `47 / 12` is 3.92, but you cannot hire
0.92 of a bus.

```python
students = 47
seats_per_bus = 12
buses_needed = math.ceil(students / seats_per_bus)
print("Buses needed:", buses_needed)
```

`help()` (Day 21) works on anything inside a module:

```python
help(math.ceil)
```

## 3. `from … import …` and `import … as …`

`from math import ceil, sqrt` copies just those names into your code, so you can skip the `math.` prefix.

```python
from math import ceil, sqrt

print(ceil(9.01))
print(sqrt(144))
```

`as` gives a module a shorter nickname. In Week 9 you will see `import numpy as np` and `import pandas as pd`
everywhere. Here is the same idea with the `statistics` module:

```python
import statistics as stats

marks = [72, 85, 64, 90, 85]
print("Mean:", stats.mean(marks))
print("Median:", stats.median(marks))
```

| Style | You write | Good when |
|-------|-----------|-----------|
| `import math` | `math.ceil(x)` | default choice: it is clear where `ceil` comes from |
| `from math import ceil` | `ceil(x)` | you use one or two names many times |
| `import statistics as stats` | `stats.mean(x)` | long module names, or a community convention (`np`, `pd`) |

Avoid `from math import *`: it dumps every name into your code and you lose track of where things come from.

## 4. A short tour of the standard library

### 4a. `random` (recap from Day 14)

`random.seed(...)` makes the "random" results repeat exactly, which is useful for demos and tests.
`random.sample(list, k)` picks `k` **different** items.

```python
import random

random.seed(3)
names = ["Asha", "Bilal", "Sara", "Omar", "Mei", "Diego"]
print("Today's presenters:", random.sample(names, 2))
```

### 4b. `datetime`: working with dates

The `datetime` module contains a type called `date`. We usually import just that name.

```python
from datetime import date

today = date.today()
print("Today is", today)
print(today.year, today.month, today.day)
```

You can make any date yourself with `date(year, month, day)`, or read one from text in `YYYY-MM-DD` form with
`date.fromisoformat(...)`.

```python
exam_day = date(2026, 12, 18)
course_start = date.fromisoformat("2026-10-12")
print(exam_day, course_start)
```

🤔 **Predict:** what do you get when you subtract one date from another? Week 6 starts on 16 November 2026.

```python
week6_start = date(2026, 11, 16)
gap = exam_day - week6_start
print(gap)
print(gap.days, "days to revise")
```

Subtracting dates gives a *time difference* object. Its `.days` is a normal `int` you can calculate with.
Dates also compare with `<` and `>`, and turn back into text with `str()` or `.isoformat()`.

```python
print(course_start < exam_day)
print(exam_day.isoformat())
```

```python
# Not a real date, so Python refuses to create it
date.fromisoformat("2026-02-30")
```

### 4c. `collections.Counter`: counting made easy

On Day 12 we counted with a dict and a loop. `Counter` does the same job in one line.

```python
from collections import Counter

votes = ["museum", "beach", "museum", "zoo", "beach", "museum"]

# The Day 12 way
counts = {}
for place in votes:
    counts[place] = counts.get(place, 0) + 1
print(counts)

# The Counter way
vote_counts = Counter(votes)
print(vote_counts)
```

🤔 **Predict:** what do these two lines show? (The second asks for a place nobody voted for.)

```python
print(vote_counts.most_common(1))
print(vote_counts["park"])
```

`most_common(n)` returns a list of `(item, count)` tuples, biggest first. A missing key gives `0` instead of a
`KeyError`.

## 6. `__name__`: was this file run, or imported?

Every module has a hidden variable `__name__`.

- In the file you **run** (`uv run something.py`, or this notebook), `__name__` is `"__main__"`.
- In a file that is **imported**, `__name__` is the module's own name.

```python
print(__name__)
```

`name_demo.py` prints its own `__name__`. 🤔 **Predict:** which of its two messages will we see?

```python
import name_demo
```

In the terminal, inside this folder, the instructor now runs `uv run name_demo.py`: there `__name__` is
`"__main__"` and the other message appears.

🤔 **Predict:** what happens if we import it a second time?

```python
import name_demo
print("done")
```

Nothing is printed by `name_demo`. Python runs a module **only the first time** it is imported and then reuses
it. That has an important consequence:

> ⚠️ If you **edit** a `.py` file after importing it, the notebook keeps using the **old** version.
> Fix: **Restart** the kernel, then run the cells again.

This is why `textutils.py` puts its self-checks under `if __name__ == "__main__":`. They run when you test the
file with `uv run textutils.py`, but stay silent when the notebook imports it.

## 7. Common mistakes (these cells fail **on purpose**)

```python
# Mistake 1: we imported ceil and sqrt from math, but not floor
floor(7.9)
```

```python
# Mistake 2: with "import statistics as stats", only the nickname exists
statistics.median([3, 1, 2])
```

```python
# Mistake 3: a typo in the module name (the file is textutils.py)
import textutil
```

| Error | Usual cause | Fix |
|-------|-------------|-----|
| `NameError: name 'floor' is not defined` | used a name you did not import | `math.floor(...)` or `from math import floor` |
| `NameError: name 'statistics' is not defined` | used the full name after `as` | use the nickname `stats` |
| `ModuleNotFoundError` | typo, or the file is in a different folder | check the spelling and the folder |

**Never name your own file after a standard module** (`random.py`, `math.py`, `datetime.py`). Python would
import *your* file instead and the real module "breaks".

## 8. Real-world example: planning a club trip

Four modules working together: `datetime`, `collections`, `random` and our own `textutils`.

```python
from collections import Counter
from datetime import date
import random

import textutils

sign_ups = ["Asha Khan", "Omar Ali", "Mei Lin", "Asha Khan", "Diego Ruiz", "Zara Shah", "Omar Ali"]
trip_day = date(2026, 12, 5)
planning_day = date(2026, 11, 16)

counts = Counter(sign_ups)
duplicates = [name for name, count in counts.items() if count > 1]
unique_names = list(counts.keys())

random.seed(1)
random.shuffle(unique_names)

print("Days until the trip:", (trip_day - planning_day).days)
print("Signed up twice:", duplicates)
print("Bus seating order:", [textutils.initials(name) for name in unique_names])
```
