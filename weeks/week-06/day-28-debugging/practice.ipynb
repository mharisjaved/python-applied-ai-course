**Bug 3.** Count how many marks are at or above the pass mark.

````python
def count_passed(marks, pass_mark=50):
    """Return how many marks are at or above pass_mark."""
    for mark in marks:
        count = 0
        if mark >= pass_mark:
            count += 1
    return count


print(count_passed([45, 50, 72, 38, 90]) == 3)
print(count_passed([60, 70]) == 2)
print(count_passed([30, 40]) == 0)
````

**Bug 4.** A valid score is between 0 and 100, including both 0 and 100.

````python
def is_valid_score(score):
    """Return True if score is between 0 and 100 (inclusive)."""
    return score >= 0 or score <= 100


print(is_valid_score(75) == True)
print(is_valid_score(0) == True)
print(is_valid_score(100) == True)
print(is_valid_score(150) == False)
print(is_valid_score(-5) == False)
````

## Exercise 4: Print debugging (bug 5)

`top_scorer` should return the name with the highest score. It returns the wrong name, and by reading alone the
bug is hard to see. Add **one** debug line inside the loop, e.g. `print(f"DEBUG {name=} {score=} {best_score=}")`,
run it, and study the output. Then fix the function and remove the debug line.

Hint: look closely at how the values are displayed.

````python
def top_scorer(results):
    """results looks like "Asha:9,Bilal:10,Sara:4". Return the name with the highest score."""
    best_name = ""
    best_score = "0"
    for part in results.split(","):
        name, score = part.split(":")
        if score > best_score:
            best_name = name
            best_score = score
    return best_name


print(top_scorer("Asha:9,Bilal:10,Sara:4"))
print(top_scorer("Asha:9,Bilal:10,Sara:4") == "Bilal")
print(top_scorer("Omar:7,Mei:12") == "Mei")
````

## Exercise 5 (challenge): Two bugs, one function

`split_bill` has **two** bugs: one gives a wrong answer, the other gives the wrong *kind* of error.
Read the docstring carefully, since it is the specification.

Try VS Code's debugger on this cell: put a breakpoint on the `tip = ...` line, open the ▷ menu next to the
cell and choose **Debug Cell**, then Step Over (F10) and watch `tip` and `share` in the Variables panel.

````python
def split_bill(total, people, tip_percent=10):
    """Return each person's share of total plus tip, rounded to 2 decimals.

    people must be at least 1, otherwise raise ValueError("people must be at least 1").
    """
    tip = total * tip_percent / 100
    share = total + tip / people
    return round(share, 2)


print(split_bill(90, 3) == 33.0)
print(split_bill(100, 4, tip_percent=20) == 30.0)
print(split_bill(50, 1, tip_percent=0) == 50.0)
````
