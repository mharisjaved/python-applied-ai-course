## 1. Indexing: one character by its position

Every character in a string has a **position number** (an *index*). Counting starts at **0**.
Negative numbers count from the end.

```
 P    Y    T    H    O    N
 0    1    2    3    4    5
-6   -5   -4   -3   -2   -1
```

````python
word = "PYTHON"
print(word[0])
print(word[3])
````

🤔 **Predict:** what is `word[-1]`?

````python
word[-1]
````

`[-1]` is always the **last** character, however long the string is.

Real use: initials from a name.

````python
first_name = "Fatima"
last_name = "Khan"
print(f"Initials: {first_name[0]}.{last_name[0]}.")
````

🤔 **Predict:** a ticket code is `"TKT-58213"`. What type is its last character, and is that digit odd?

````python
ticket = "TKT-58213"
last_char = ticket[-1]
print(last_char, type(last_char))
print(int(last_char) % 2)     # Day 03: remainder 1 means odd
````

A character taken from a string is still a `str`. Convert it before doing maths.

🤔 **Predict:** `word` has 6 characters. What happens with `word[6]`?

````python
word[6]
````

`IndexError`: there is no position 6. The last index is always `len(word) - 1`, which is 5 here.

## 2. Slicing: a piece of a string

`text[start:stop]` takes the characters **from `start` up to, but not including, `stop`**.
Think of `stop` as a fence: you stop *before* it. The slice has `stop - start` characters.

````python
print(word)

print(word[0:3])
print(word[2:5])
````

Leave out `start` to begin at the start, leave out `stop` to go to the end.

````python
print(word[:2])
print(word[2:])
````

Real use: a date always has the same layout `YYYY-MM-DD`, so slices can pull out each part.

````python
date_text = "2026-10-16"
year = date_text[:4]
month = date_text[5:7]
day = date_text[8:]
print(f"Year {year}, month {month}, day {day}")
print(int(year) + 1)
````

Negative numbers work in slices too: "the last 3 characters", "everything except the last 4".

````python
file_name = "sales_report.csv"
print(file_name[-3:])
print(file_name[:-4])
````

🤔 **Predict:** `word[0:100]`, an error or not?

````python
word[0:100]
````

Slices are forgiving: they stop at the end of the string. A single index is strict and raises `IndexError`.

## 3. Step and reversing

A third number is the **step**: `text[start:stop:step]`.

````python
print(word[2:5:2])      # every 2nd character
print(word[::-1])     # step -1 walks backwards: reversed
````

🤔 **Predict:**

````python
"stressed"[::-1]
````

## 4. Strings still never change

````python
word[0] = "J"
````

Same idea as Day 04: strings are immutable. Build a **new** string instead:

````python
new_word = "J" + word[1:]
print(new_word)
````

## 5. Formatting numbers in f-strings

After a variable inside `{ }`, a colon `:` starts a **format spec**: instructions for how to show the value.

| Spec | Meaning | `f"{1234.5:…}"` |
|------|---------|-----------------|
| `.2f` | 2 decimal places | `1234.50` |
| `,` | thousands separator | `1,234.5` |
| `,.2f` | both | `1,234.50` |

````python
price = 1234.5102384
print(f"{price}")
print(f"{price:.3f}")
print(f"{price:,}")
print(f"{price:,.2f}")
````

🤔 **Predict:** remember Day 03's bill splitter?

````python
share = 26.546666666666667
print(f"Each person pays {share:.2f}")
print(share)
````

`:.2f` rounds **for display only**; `share` itself is unchanged.

**Width and alignment:** `:<10` = left-aligned in 10 spaces, `:>10` = right-aligned.
The square brackets only make the spaces visible.

````python
item = "Tea"
print(f"[{item:<10}]")
print(f"[{item:>10}]")
print(f"[{4.5:>8.2f}]")
````

Numbers line up when they are right-aligned with the same width and decimals.
A plain text value can go inside the braces too: use the *other* kind of quotes (`'Tea'` inside `f"..."`).

````python
print(f"{'Tea':<12}{4.5:>8.2f}")
print(f"{'Sandwich':<12}{12.75:>8.2f}")
print(f"{'Water':<12}{1:>8.2f}")
````

## 6. Common mistakes

````python
# Mistake 1: the number is still text (Day 03), and text has no 2-decimal format
price_text = "4.5"
print(f"{price_text:.2f}")
````

`ValueError: Unknown format code 'f' for object of type 'str'`. Fix: `float(price_text)` first.

These two give **no error**, just the wrong output:

````python
print(date_text[0:3])     # wanted the year: stop is NOT included, so use [0:4]
print(f"{price:.2}")      # forgot the f: .2 alone means "2 significant digits"
````

## 7. Real-world example: a formatted receipt

````python
item1 = "Laptop"
qty1 = 1
price1 = 1249.00
item2 = "Wireless mouse"
qty2 = 2
price2 = 18.50
item3 = "USB-C cable"
qty3 = 3
price3 = 4.99

line1 = qty1 * price1
line2 = qty2 * price2
line3 = qty3 * price3
subtotal = line1 + line2 + line3
tax = subtotal * 0.08
total = subtotal + tax
````

````python
print("=" * 32)
print(f"{'Item':<16}{'Qty':>4}{'Amount':>12}")
print("-" * 32)
print(f"{item1:<16}{qty1:>4}{line1:>12,.2f}")
print(f"{item2:<16}{qty2:>4}{line2:>12,.2f}")
print(f"{item3:<16}{qty3:>4}{line3:>12,.2f}")
print("-" * 32)
print(f"{'Subtotal':<20}{subtotal:>12,.2f}")
print(f"{'Tax (8%)':<20}{tax:>12,.2f}")
print(f"{'TOTAL':<20}{total:>12,.2f}")
print("=" * 32)
````

Every row adds up to 32 characters (16 + 4 + 12, or 20 + 12), so the columns line up.
Three items needed nine variables. Next week, **lists** will hold all of them in one variable.

## Recap

1. `text[0]` is the first character, `text[-1]` the last. Too far → `IndexError`.
2. `text[start:stop]` stops **before** `stop`. Leave out either end. `text[::-1]` reverses.
3. Strings cannot be changed in place. Build a new one: `"J" + word[1:]`.
4. Format specs: `{x:.2f}` 2 decimals, `{x:,}` thousands, `{x:<10}` left, `{x:>10}` right.
