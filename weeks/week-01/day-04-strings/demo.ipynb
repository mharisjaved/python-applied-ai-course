## 1. Making strings: quotes, `\n` and `\t`

Single and double quotes make exactly the same kind of string. Pick the one that does not clash with the text.

```python
greeting = 'Hello'
place = "Mei's cafe"      # double quotes, so the apostrophe is fine
print(greeting, place)
```

A backslash starts an **escape sequence**: `\n` means "new line", `\t` means "tab".

🤔 **Predict:** how many lines will this print?

```python
print("Name:\tAsha\nCity:\tLahore")
```

Triple quotes (`"""` or `'''`) let one string span several lines.

```python
menu = """--- Cafe menu ---
1. Tea
2. Coffee
3. Mango juice"""
print(menu)
```

## 2. Joining, repeating and measuring

`+` joins two strings, `*` repeats a string, `len()` counts characters.

```python
first_name = "Mei"
last_name = "Lin"
print(first_name + last_name) # concatination
```

🤔 **Predict:** how do we get a space between the names?

```python
full_name = first_name + " " + last_name
print(full_name)
```

```python
print("-" * 20)
print("ha" * 3)

# import sys
# sys.getsizeof(int)
```

Remember Day 03: `"20" * 2` gave `'2020'`. Now you know why: text times a number **repeats** the text.

🤔 **Predict:** `len(full_name)` is 6 or 7?

```python
len(full_name)
```

Spaces are characters too, and `len()` counts them.

## 3. Methods that change text

A **method** is a tool that belongs to a value. Write the value, a **dot**, the method name and **parentheses**.

```python
city = "  kArAcHi "
print(city.strip())     # remove spaces at both ends
print(city.upper())     # ALL CAPITALS
print(city.lower())     # all lowercase
print(city.title())     # First Letter Of Each Word
```

`.replace(old, new)` swaps every `old` piece for `new`.

```python
phone = "0300-123-4567"
print(phone.replace("-", ""))
print(phone.replace("-", " "))
```

Methods can be **chained**. They run left to right: first `strip`, then `title`.

```python
print(city.strip().title())
```

## 4. Methods that search text

```python
sentence = "the cat sat on the mat"
print(sentence.count("at"))     # how many times?
print(sentence.count("the"))
```

`.find()` gives the **position** where the text starts. Positions count from **0** (more about that tomorrow).
If the text is not there, you get `-1` (not an error).

```python
print(sentence.find("sat"))
print(sentence.find("dog"))
```

`.startswith()` and `.endswith()` answer yes/no questions, so they give a `bool`.

```python
email = "omar@example.com"
file_name = "report_october.pdf"
print(email.endswith(".com"))
print(file_name.startswith("report"))
print(file_name.endswith(".docx"))
```

🤔 **Predict:** how many letter p's does Python find in `"Python"`?

```python
"Python".count("p")
```

Strings are **case-sensitive**: `"P"` and `"p"` are different characters. Lowercase first, then search:

```python
"Python".lower().count("p")
```

## 5. Strings never change

🤔 **Predict:** what does the last line print?

```python
name = "zara"
name.upper()
print(name)
```

`name.upper()` made a **new** string `'ZARA'` and threw it away, because we did not keep it.
The original string cannot be changed: strings are **immutable**.

To keep the result, attach a name to it (Day 02: right side first):

```python
name = name.upper()
print(name)
```

## 6. f-strings: values inside text

Building messages with commas or `+` gets clumsy:

```python
name = "Zara"
total = 26.55
print("Hello,", name, "!")              # unwanted space before !
print("Total: " + str(total) + " each")   # needs str() and lots of +
```

Put an **`f`** before the opening quote, then write variables inside **`{ }`**:

```python
print(f"Hello, {name}!")
print(f"Total: {total} each")
```

Any expression can go inside the braces, including maths and methods:

```python
item = "notebook"
quantity = 3
price = 2.5
print(f"{quantity} x {item} = {quantity * price}")
print(f"{item.upper()} has {len(item)} letters")
```

An f-string is just a string, so it can be stored in a variable:

```python
receipt_line = f"{quantity} x {item}"
print(receipt_line)
print(len(receipt_line))
```

## 8. Real-world example: username and email generator

```python
full_name = input("Full name: ")

display_name = full_name.strip().title()
username = full_name.strip().lower().replace(" ", ".")
email = f"{username}@student.example.com"

print(f"Welcome, {display_name}!")
print(f"Username: {username}")
print(f"Email: {email}")
print(f"Your username has {len(username)} characters")
```

One messy input, three clean outputs, and the original `full_name` is unchanged.
