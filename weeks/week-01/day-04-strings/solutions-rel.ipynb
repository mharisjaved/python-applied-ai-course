## Exercise 1: Predict, then check

| Expression | Result | Why |
|------------|--------|-----|
| `len("Hello World")` | `11` | the space counts |
| `"ha" * 3` | `hahaha` | `*` repeats |
| `"Data" + "Science"` | `DataScience` | `+` adds no space |
| `"  hi  ".strip()` | `hi` | |
| `"banana".count("a")` | `3` | |
| `"banana".find("n")` | `2` | positions start at 0: b=0, a=1, n=2 |
| `"banana".find("z")` | `-1` | not found is `-1`, not an error |
| `"good MORNING".title()` | `Good Morning` | `.title()` also lowercases the rest |
| `"report.pdf".endswith(".pdf")` | `True` | |
| `"Python".startswith("p")` | `False` | case-sensitive |

Most common wrong predictions: `10` for `len("Hello World")`, `3` for `find("n")` (counting from 1),
an error for `find("z")`, and `True` for `startswith("p")`.

```python
print(len("Hello World"))
print("ha" * 3)
print("Data" + "Science")
print("  hi  ".strip())
print("banana".count("a"))
print("banana".find("n"))
print("banana".find("z"))
print("good MORNING".title())
print("report.pdf".endswith(".pdf"))
print("Python".startswith("p"))
```

## Exercise 2: Clean up messy input

Chain the methods: `strip` first, then change the case. The phone only needs `.replace()`.

```python
raw_city = "   lAHORE  "                   # pretend the user typed this
raw_email = "  Sara.Ahmed@Example.COM "     # pretend the user typed this
raw_phone = "0300 123 4567"                 # pretend the user typed this

city = raw_city.strip().title()
email = raw_email.strip().lower()
phone = raw_phone.replace(" ", "")

print(f"City: {city}")
print(f"Email: {email}")
print(f"Phone: {phone}")
print(f"Email length: {len(raw_email)} before, {len(email)} after")
```

Watch for: `.capitalize()` found online (it also works for one word; fine), and students who print
`raw_email.strip()` but then measure `len(raw_email)` twice because they never stored the cleaned value.

```python
# self-check
print(city == "Lahore", email == "sara.ahmed@example.com", phone == "03001234567")
```

## Exercise 4: Review inspector

```python
review = "  The food was GREAT but the delivery was terrible. Great price though!  "

clean_review = review.strip().lower()

print(f"Characters: {len(clean_review)}")
print(f"'great' appears {clean_review.count('great')} times")
print(f"'delivery' starts at position {clean_review.find('delivery')}")
print(f"Ends with '!': {clean_review.endswith('!')}")
print(f"Polite version: {clean_review.replace('terrible', 'slow')}")
```

Reasoning to share: we lowercase **before** counting because `"GREAT"`, `"Great"` and `"great"` are different
strings to Python; on the raw review `review.count("great")` gives `0`. We strip before `.endswith('!')`
because the raw review ends with spaces, so `review.endswith("!")` is `False`.

Watch for: quotes inside f-strings. `f"'great' appears {clean_review.count('great')} times"` uses single
quotes inside double quotes. If students get a `SyntaxError` here, the simplest fix is to compute the value in
a variable first:

```python
great_count = clean_review.count("great")
print(f"'great' appears {great_count} times")
print(review.count("great"), review.endswith("!"))   # the raw review gives 0 and False
```

## Exercise 5: Username and email generator

Clean once, then reuse. Building each value in its own variable keeps the f-strings short.

```python
full_name = "  zara AHMED  "   # pretend the user typed this
birth_year = 2004

clean_name = full_name.strip()
display_name = clean_name.title()
username = clean_name.lower().replace(" ", ".")
email = f"{username}@student.example.com"
gamer_tag = clean_name.upper().replace(" ", "_") + "_" + str(birth_year)

print(f"Display name: {display_name}")
print(f"Username: {username}")
print(f"Email: {email}")
print(f"Gamer tag: {gamer_tag}")
print(f"Username length: {len(username)}")
```

```python
# self-check
print(username == "zara.ahmed", gamer_tag == "ZARA_AHMED_2004")
```

Common mistakes:
- `replace` before `strip`: `"  zara AHMED  ".replace(" ", ".")` gives `..zara.AHMED..`. Order matters.
- `gamer_tag = ... + "_" + birth_year` → `TypeError`. Use `str(birth_year)` or an f-string:
  `f"{clean_name.upper().replace(' ', '_')}_{birth_year}"`.
- A name typed with **two** spaces in the middle (`"zara  ahmed"`) gives `zara..ahmed`. Fixing that properly
  needs `.split()` (Day 07). Mention it as a known limitation.

**Step 2** with `input()` (mock answer: `"  diego RAMOS "`):

```python
full_name = input("Full name: ")

clean_name = full_name.strip()
username = clean_name.lower().replace(" ", ".")
print(f"Display name: {clean_name.title()}")
print(f"Username: {username}")
print(f"Email: {username}@student.example.com")
```

## Exercise 6 (challenge): Name badge

The border is 4 characters wider than the name: `"* "` on the left and `" *"` on the right.

```python
badge_name = "Zara Ahmed"

border = "*" * (len(badge_name) + 4)
print(border)
print(f"* {badge_name} *")
print(border)
```

Common wrong answer: `"*" * len(badge_name) + 4` → `TypeError` (Day 03 precedence: `*` happens first, then
text + number). The parentheses are needed.

**Bonus:** a single `print()` using `\n`:

```python
badge_name = "Fatima Zahra Khan"

border = "*" * (len(badge_name) + 4)
print(f"{border}\n* {badge_name} *\n{border}")
```
