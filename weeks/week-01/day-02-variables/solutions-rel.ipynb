## Exercise 4: Trace the score

Predictions: `5`, then `15`, then `0 15`.

| Line | `points` | `bonus` |
|------|----------|---------|
| `points = 5` | 5 | – |
| `points = points + 10` | 15 | – |
| `bonus = points` | 15 | 15 |
| `points = 0` | 0 | 15 |

The key idea: `bonus = points` copies the *current value*; changing `points` later does not affect `bonus`.

````python
points = 5
print(points)
points = points + 10
print(points)
bonus = points
points = 0
print(points, bonus)
````

## Exercise 5: Swap two values

Use a temporary variable (the "third cup").

````python
cup_a = "tea"
cup_b = "coffee"

temp = cup_a      # put tea in the spare cup
cup_a = cup_b     # cup_a now holds coffee
cup_b = temp      # cup_b gets the tea from the spare cup

print("cup_a:", cup_a)
print("cup_b:", cup_b)
````

Common wrong attempt, worth showing on the board:

```python
cup_a = cup_b   # both are now "coffee", and the tea is lost!
cup_b = cup_a
```

**Alternative (Python shortcut):** `cup_a, cup_b = cup_b, cup_a`. Show it only *after* students understand the
temporary-variable version; tuple unpacking is formally taught on Day 07.

````python
cup_a = "tea"
cup_b = "coffee"
cup_a, cup_b = cup_b, cup_a
print("cup_a:", cup_a)
print("cup_b:", cup_b)
````

## Exercise 6 (challenge): Library card

The student ID must be a **`str`**: it is an identifier, not a quantity (we never do maths with it), and as an
`int` the leading zeros would be lost. In fact `00731` without quotes is a `SyntaxError` in Python.
Same rule later for phone numbers and postcodes.

````python
book_title = "Python for Beginners"
page_count = 320
late_fee_per_day = 0.5
is_borrowed = True
student_id = "00731"

print("--- Library Card ---")
print("Title:", book_title)
print("Pages:", page_count)
print("Late fee per day:", late_fee_per_day)
print("Borrowed:", is_borrowed)
print("Student ID:", student_id)
````
