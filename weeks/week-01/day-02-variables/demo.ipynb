## 1. Variables: name tags for values

A variable is a **name tag** attached to a value. `=` means *"attach this name to this value"*.
Read it **right to left**: take `"Lahore"`, attach the name `city`.

```python
# x = 10
city = "Lahore"
print(city)
```

Assignment itself shows nothing. Use the name to get the value back.
`print()` can show several values. Separate them with commas and it adds a space between them.

```python
# print("I live in city")

print("I live in", city)
# print("I live in "+ city)
```

🤔 **Predict:** what does this show: the word `city` or `Lahore`?

```python
print("city")
print(city)
```

Quotes mean *literal text*. Without quotes, Python looks up a variable with that name.

## 2. The four basic data types

| Type | Meaning | Example |
|------|---------|---------|
| `int` | whole number | `21` |
| `float` | number with a decimal point | `1.74` |
| `str` | text, always in quotes | `"Asha"` |
| `bool` | yes / no | `True` or `False` |

```python
students_in_class = 24        # int
price = 149.99                # float
course_name = "Python + AI"   # str
is_online = False             # bool

print(students_in_class, price, course_name, is_online)
```

`type()` answers the question *"what kind of value is this?"*

```python
print(type(students_in_class))
print(type(price))
print(type(course_name))
print(type(is_online))
```

```python
type(False)
```

- `21` is a number but `"21"` is text: the quotes decide.
- `3.0` is a `float` because of the decimal point.
- `"False"` is a `str`. The real boolean is `False`: capital F, no quotes.

## 3. Reassignment: moving the name tag

```python
score = 10
print("Start:", score)

score = 25
print("After reassignment:", score)
```

🤔 **Predict:** what is `score` after the next cell? Remember: **right side first**.

```python
score = score + 5
print(score)
```

Trace it like this:

| Line | `score` |
|------|---------|
| `score = 10` | 10 |
| `score = 25` | 25 |
| `score = score + 5` | 25 + 5 → **30** |

🤔 **Predict** both values before running:

```python
a = 7
b = a
a = 100
print(a, b)
```

`b = a` attached `b` to the value `a` had **at that moment** (7). Moving `a` later does not move `b`.
