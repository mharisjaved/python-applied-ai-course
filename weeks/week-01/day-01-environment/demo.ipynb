## 1. The first program

`print()` shows whatever is inside the brackets. Text goes inside quotes.

````python
print("Hello, World!")
````

🤔 **Predict:** what will this show?

````python
print("My name is Asha")
print("I am learning Python")

print("1 A, 2 B")
print(1, "A", 2, "B")
# print(Saim)
````

The interpreter runs instructions **top to bottom**, one at a time, so the lines appear in order.

## 2. Printing numbers

Numbers do not need quotes.

````python
print(2026)
print(3.5)
````

## Recap

- A program is a list of instructions; the interpreter runs them top to bottom.
- `print("text")` shows text on the screen; quotes mark text.
- Python is case-sensitive, and every opening quote/bracket needs a closing one.
- In our setup, `uv run hello.py` runs a file.
