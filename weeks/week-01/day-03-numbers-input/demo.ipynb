## 2. Floor division `//` and remainder `%`

17 sweets shared fairly between 5 children: each child gets **3**, and **2** are left over.

```python
sweets = 17
children = 5
print("Exact division:", sweets / children)
print("Each child gets:", sweets // children)
print("Left over:", sweets % children)
```

- `//` answers *"how many **whole** times does it fit?"*
- `%` answers *"what is **left over**?"* (`%` is **not** percent!)

🤔 **Predict:** a bus journey takes 200 minutes. How many whole hours is that, and how many minutes are left over?

```python
journey_minutes = 200
hours = journey_minutes // 60
minutes = journey_minutes % 60
print(journey_minutes, "minutes =", hours, "hours and", minutes, "minutes")
```

**Even or odd?** Divide by 2 and look at the remainder: `0` means even, `1` means odd.

```python
print(14 % 2) # Even
print(27 % 2) # remainder : 1 (odd)
```

## 3. Order of operations

🤔 **Predict:** Asha scored 78, 85 and 92 in three quizzes. Is this her average?

```python
78 + 85 + 92 / 3
```

Python does `/` before `+`, so only 92 was divided by 3. Parentheses fix it:

```python
average = (78 + 85 + 92) / 3
print("Average:", average)
```

| Order | Operators |
|-------|-----------|
| 1st | `( )` parentheses |
| 2nd | `**` |
| 3rd | `*` `/` `//` `%` |
| 4th | `+` `-` |

Operators on the same level run left to right. **When in doubt, add parentheses.**

🤔 **Predict both:**

```python
print(2 + 3 * 4)
print(10 - 4 - 3)
```

## 4. Updating a variable: `+=`

On Day 02 we wrote `score = score + 5`. Updating a running total is so common that Python has a shortcut.

```python
cart_total = 0
cart_total += 4.5    # same as: cart_total = cart_total + 4.5
cart_total += 12.0
cart_total += 3.25
print("Cart total:", cart_total)
```

```python
wallet = 50
wallet -= 19.75      # same as: wallet = wallet - 19.75
print("Wallet:", wallet)
```

The same shortcut works with the other operators: `*=`, `/=`, `//=`, `%=`.
Careful: `=+` is **not** the same as `+=` (see section 8).

## 6. Converting types: `int()`, `float()`, `str()`

```python
age = int(age_text)
print(type(age))
print("Next year you will be", age + 1)
```

| Call | Result | What it does |
|------|--------|--------------|
| `int("42")` | `42` | text → whole number |
| `float("4.75")` | `4.75` | text → decimal number |
| `str(2026)` | `'2026'` | number → text |
| `int(9.99)` | `9` | cuts off the decimals, does **not** round |

```python
print(float("4.75") * 2)
print(int(9.99))
print(str(2026), type(str(2026)))
```

Usually we convert straight away. Read it **inside out**: `input()` runs first, then `int()` converts its answer.

```python
pages = int(input("How many pages did you read today? "))
print("At this pace you read", pages * 7, "pages a week")
```

## 7. Rounding: `round()`

🤔 **Predict:**

```python
round(1.134245326,2)
```

Computers store most decimals *approximately* (like writing 1/3 as 0.333… on paper), so tiny errors appear.
`round(number, digits)` tidies a result for display.

```python
print(round(0.1 + 0.2, 2))
print(round(85.6666, 1))
print(round(7.8))
```

- `round(x, 2)` → 2 decimal places (money)
- `round(x)` → nearest whole number (an `int`)
- Round **at the end**, when you show the result, not in the middle of a calculation.

maths → `int("5") + 5` gives `10`; text → `"5" + str(5)` gives `'55'`.

```python
# Mistake 3: int() cannot read a decimal point inside text
int("3.5")

# int(float("3.5"))
```

`ValueError`: the *type* is fine (it is a `str`) but this *value* cannot be converted.
For decimals use `float("3.5")`.

## 9. Real-world example: bill splitter

```python
bill = float(input("Total bill: "))
people = int(input("Number of people: "))
tip_percent = float(input("Tip percent: "))

tip = bill * tip_percent / 100
total = bill + tip
share = total / people

print("Tip:", round(tip, 2))
print("Total with tip:", round(total, 2))
print("Each person pays:", round(share, 2))
```

Notice: `float` for money, `int` for a number of people, and `tip_percent / 100` because `%` is not percent.
Try `print(share)` without `round` to see why we round.

## Recap

1. `+ - * / **` work like a calculator. `/` always gives a `float`.
2. `//` = how many whole times, `%` = what is left over (minutes → hours, even/odd).
3. `**` first, then `* / // %`, then `+ -`. Parentheses make your intention clear.
4. `x += 5` is short for `x = x + 5`.
5. `input()` always returns a `str`. Convert with `int()` or `float()` before doing maths.
6. `TypeError`: wrong types combined. `ValueError`: text that cannot be converted.
