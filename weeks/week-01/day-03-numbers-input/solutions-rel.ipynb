## Exercise 1: Predict, then check

| Expression | Result | Why |
|------------|--------|-----|
| `7 + 3 * 2` | `13` | `*` before `+` |
| `(7 + 3) * 2` | `20` | parentheses first |
| `20 / 4` | `5.0` | `/` always gives a `float` |
| `9 / 2` | `4.5` | |
| `9 // 2` | `4` | whole times only |
| `20 % 6` | `2` | 6 fits 3 times (18), 2 left over |
| `2 ** 4` | `16` | 2 × 2 × 2 × 2 |
| `10 - 4 - 3` | `3` | left to right: (10 − 4) − 3 |
| `int("7") + 3` | `10` | converted first, then added |
| `"7" + "3"` | `73` | two texts are glued together |
| `int(4.9)` | `4` | `int()` cuts off decimals, it does not round |
| `type(20 / 4)` | `<class 'float'>` | |

Most common wrong predictions: `5` for `20 / 4`, `10` for `"7" + "3"`, `5` for `int(4.9)`, and `5` for `10 - 4 - 3`
(students who think right-to-left).

```python
print(7 + 3 * 2)
print((7 + 3) * 2)
print(20 / 4)
print(9 / 2)
print(9 // 2)
print(20 % 6)
print(2 ** 4)
print(10 - 4 - 3)
print(int("7") + 3)
print("7" + "3")
print(int(4.9))
print(type(20 / 4))
```

## Exercise 2: Hours, pairs and leftovers

**a)** `//` gives the whole hours and `%` gives the minutes left over. Because the code only uses the variable,
it works for 90 (`1 hours and 30 minutes`) and 200 (`3 hours and 20 minutes`) unchanged.
"1 hours" is grammatically clumsy; fixing it needs `if` (Day 09). Mention that, do not solve it now.

```python
film_minutes = 135

hours = film_minutes // 60
minutes = film_minutes % 60
print(film_minutes, "minutes =", hours, "hours and", minutes, "minutes")
```

```python
# self-check
print(hours == 2, minutes == 15)
```

Common wrong answer: `hours = film_minutes / 60` → `2.25`, then students try to "turn 0.25 into minutes".
Ask: *"How many **whole** hours?"* and point them back to `//`.

**b)** `27 // 2` pairs and `27 % 2` left over. A remainder of `1` when dividing by 2 means **odd**.

```python
students = 27

print("Pairs:", students // 2)
print("Without a partner:", students % 2)
```

## Exercise 4: Temperature converter

**a)** Convert the text with `float()` (not `int()`: the value has a decimal point).

```python
celsius_text = "31.5"   # pretend the user typed this

celsius = float(celsius_text)
fahrenheit = celsius * 9 / 5 + 32
print(celsius, "C =", fahrenheit, "F")
```

**b)** The parentheses are essential. Without them Python computes `75 - (32 * 5 / 9)` = `57.22…`.

```python
fahrenheit = 75

celsius = (fahrenheit - 32) * 5 / 9
print(fahrenheit, "F =", round(celsius, 1), "C")
```

```python
# self-check, and the most common wrong version for comparison
print(round(celsius, 1) == 23.9)
print("Without parentheses:", fahrenheit - 32 * 5 / 9)
```

## Exercise 5: Bill splitter

Convert each text value once, at the top, and give the numbers clear names. Round only when printing.

```python
bill_text = "135.80"      # pretend the user typed this
people_text = "4"         # pretend the user typed this
tip_percent_text = "15"   # pretend the user typed this

bill = float(bill_text)
people = int(people_text)
tip_percent = float(tip_percent_text)

tip = bill * tip_percent / 100
total = bill + tip
share = total / people

print("Bill:", bill)
print("Tip:", round(tip, 2))
print("Total:", round(total, 2))
print("Each of", people, "people pays:", round(share, 2))
```

```python
# self-check
print(round(share, 2) == 39.04)
print("Unrounded total:", total)   # 156.17000000000002: this is why we round
```

Points to raise:
- `Bill: 135.8` drops the trailing zero. Students will ask; Day 05 fixes it with `:.2f`.
- Common bug: `tip = bill * 15%`. `%` is the remainder operator, not percent.
- Common bug: `people = float(people_text)` works, but `int` says "a whole number of people" more clearly.

**Step 2.** The same program with `input()` (convert directly, read inside out):

```python
bill = float(input("Total bill: "))
people = int(input("Number of people: "))
tip_percent = float(input("Tip percent: "))

tip = bill * tip_percent / 100
total = bill + tip
share = total / people

print("Bill:", bill)
print("Tip:", round(tip, 2))
print("Total:", round(total, 2))
print("Each of", people, "people pays:", round(share, 2))
```

**Step 3.** Typing `four` raises a **`ValueError`** at the `int(...)` line. The bill was read fine, so the error
appears on the second question. Preventing the crash needs `if` (Day 09) and `try/except` (Day 25).

```python
bill = float(input("Total bill: "))
people = int(input("Number of people: "))
```

## Exercise 6 (challenge): Screen-time report

Peel off the biggest unit first, then keep only what is left over with `%`.

```python
seconds_used = 200000

days = seconds_used // 86400
remaining = seconds_used % 86400     # seconds left after taking out whole days

hours = remaining // 3600
remaining = remaining % 3600         # seconds left after taking out whole hours

minutes = remaining // 60
seconds = remaining % 60

print(seconds_used, "seconds =", days, "days,", hours, "hours,", minutes, "minutes and", seconds, "seconds")
```

```python
# self-check
print(days == 2, hours == 7, minutes == 33, seconds == 20)
```

Common wrong answer: `hours = seconds_used // 3600` → `55` (the *total* hours, not the hours left after the days).

**Alternative** using the update shortcut `%=` and one-line expressions. Shorter, but harder to read for
beginners, so show it only after the version above works:

```python
seconds_used = 1000000

remaining = seconds_used
days = remaining // 86400
remaining %= 86400
hours = remaining // 3600
remaining %= 3600
print(days, "days,", hours, "hours,", remaining // 60, "minutes and", remaining % 60, "seconds")
```
