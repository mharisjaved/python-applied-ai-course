## Exercise 6 (challenge): Screen-time report

Zara's phone says she used a video app for **200000 seconds** this month. Convert that into days, hours, minutes
and seconds.

```
200000 seconds = 2 days, 7 hours, 33 minutes and 20 seconds
```

Useful facts: 1 minute = 60 seconds · 1 hour = 3600 seconds · 1 day = 86400 seconds.

When it works, change the value to `1000000`. You should get `11 days, 13 hours, 46 minutes and 40 seconds`.

````python
seconds_used = 200000

remaining = seconds_used
days = remaining // 86400

remaining %= 86400
hours = remaining // 3600

remaining %= 3600
minutes = remaining // 60

remaining %= 60
seconds = remaining

print(days, "days,", hours, "hours,", minutes, "minutes and", seconds, "seconds")
````
