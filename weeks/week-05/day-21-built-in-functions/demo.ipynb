## 1. Warm-up: the loop we already know

On Day 12 we found the highest mark by hand. It works, but it takes five lines.

```python
marks = [72, 85, 64, 91, 58, 79]

highest = marks[0]
for mark in marks:
    if mark > highest:
        highest = mark
print("Highest (loop):", highest)
```

```python
print(max(marks), min(marks), sum(marks), len(marks))
```

Built-ins combine into small formulas. An average is just `sum / len`:

```python
average = sum(marks) / len(marks)
print("Average:", average)
print("Rounded:", round(average, 1))
```

**Rule of thumb:** if you are about to write a loop that totals, counts, finds the biggest or the smallest,
check whether a built-in already does it.

## 2. Numbers: `abs()` and `round()`

`abs()` gives the **distance from zero**: it removes the minus sign. Useful whenever you care about
*how far off* something is, not in which direction.

```python
budget = 250
spent = 287.5

difference = budget - spent
print("Difference:", difference)
print("Off budget by:", abs(difference))
```

🤔 **Predict** each value.

```python
print(abs(-12), abs(12), abs(3 - 10))
print(round(3.14159, 2), round(2.71828), round(19.999, 1))
```

- `round(x)` with no second argument gives an `int`.
- `round(x, 2)` keeps 2 decimal places (it changes the value; `:.2f` in an f-string only changes how it is *shown*).

## 3. `sorted()` with `reverse=` and `key=`

`sorted()` returns a **new** sorted list and leaves the original alone (Day 07).

```python
scores = [64, 91, 58, 85]
print(sorted(scores))
print(sorted(scores, reverse=True))
print("Original:", scores)
```

🤔 **Predict** the order. Watch the capital letters.

```python
names = ["omar", "Zara", "asha", "Bilal", "mei"]
print(sorted(names))
```

Python compares text letter by letter using character codes, and **every capital letter comes before every
lowercase letter**. So `"Zara"` beats `"asha"`.

`key=` tells `sorted()` *what to compare*. Python calls the key function on every item, sorts by those results,
but gives back the **original** items.

```python
print(sorted(names, key=str.lower))
```

`str.lower` is the `.lower()` method itself, **without brackets**: we hand the tool to `sorted()` and it calls it
on each name (`"Zara"` → `"zara"`).

🤔 **Predict:** sorting by length.

```python
words = ["notebook", "pen", "calculator", "ruler"]
print(sorted(words, key=len))
print(sorted(words, key=len, reverse=True))
```

### Sorting dictionaries with your own key function

A list of dicts cannot be sorted directly: Python does not know whether to compare names or scores.
We write a tiny function that returns *the thing to sort by* and pass its name as `key`.

```python
students = [
    {"name": "Sara", "score": 78},
    {"name": "Ken", "score": 92},
    {"name": "Fatima", "score": 85},
]

def by_score(student):
    """Sort key: return the student's score."""
    return student["score"]

for student in sorted(students, key=by_score, reverse=True):
    print(student["name"], student["score"])
```

`max()` and `min()` accept the same `key=`:

```python
best = max(students, key=by_score)
print("Top student:", best["name"])
```

## 4. `zip()` and `enumerate()`

`zip()` walks through two lists **side by side**, like the two halves of a zip.

```python
names = ["Asha", "Bilal", "Mei"]
scores = [88, 75, 93]

for name, score in zip(names, scores):
    print(name, "scored", score)
```

On its own, `zip()` gives a "zip object" that hands out pairs one at a time. Wrap it in `list()` to see them all.

```python
print(zip(names, scores))
print(list(zip(names, scores)))
```

🤔 **Predict:** the lists have different lengths. What happens to Mei?

```python
print(list(zip(["Asha", "Bilal", "Mei"], [88, 75])))
```

`zip()` **stops at the shorter list** and gives no error. Mei silently disappears. Always check both lists have the
same length (`len()`) when the data matters.

`enumerate()` recap (Day 12): numbers items as you loop. `start=1` gives human-friendly positions.

```python
for position, name in enumerate(names, start=1):
    print(f"{position}. {name}")
```

## 5. `any()`, `all()` and checking text

`all(list)` → `True` if **every** item is truthy. `any(list)` → `True` if **at least one** is.
Remember from Day 09: an empty string `""` counts as `False`.

```python
name = "Diego"
email = ""
phone = "0300-1234567"

fields = [name, email, phone]
print("All fields filled:", all(fields))
print("At least one filled:", any(fields))
```

With a list of `True`/`False` results, `any`/`all` answer "did anyone…?" and "did everyone…?".

```python
marks = [72, 85, 64, 91, 58, 79]

passed = []
for mark in marks:
    passed.append(mark >= 60)

print(passed)
print("Everyone passed:", all(passed))
print("Someone passed:", any(passed))
```

### `.isdigit()` and `.isalpha()`

🤔 **Predict** `True` or `False` for each line.

```python
print("4821".isdigit())
print("48 21".isdigit())
print("-5".isdigit())
print("3.5".isdigit())
```

```python
print("Fatima".isalpha())
print("Mei Ling".isalpha())
print("Zara2".isalpha())
```

- `.isdigit()` means "only the characters 0–9". Good for PINs, ages, quantities. A minus sign or a decimal point
  makes it `False`, so it cannot validate prices or temperatures (Day 25 shows the proper way).
- `.isalpha()` means "only letters". A space counts as *not* a letter.

## 7. Common mistakes (these cells fail **on purpose**)

```python
# Mistake 1: calling the key function instead of passing it
sorted(words, key=len())
```

```python
# Mistake 2: same mistake with our own function
sorted(students, key=by_score())
```

```python
# Mistake 3: numbers stored as text cannot be summed
sum(["12", "30", "8"])
```

```python
# Mistake 4: a list mixing text and numbers cannot be sorted
sorted(["Asha", 7, "Mei"])
```

And one that gives **no error** but a wrong result:

```python
ranked = scores.sort()
print(ranked)
```

`.sort()` sorts the list *in place* and returns `None`. Use `sorted()` when you want the result in a new variable.

| Mistake | Fix |
|---------|-----|
| `key=len()` / `key=by_score()` | pass the function **name**, no brackets |
| `sum` of strings | convert first: `int(text)` |
| `x = my_list.sort()` | `x = sorted(my_list)` |

## 8. Real-world example: a quiz leaderboard

Scores from the Week 3 Quiz Game. One function per job, built-ins doing the heavy lifting.

```python
players = [
    {"name": "Asha", "score": 82, "seconds": 95},
    {"name": "Bilal", "score": 67, "seconds": 120},
    {"name": "Zara", "score": 91, "seconds": 110},
    {"name": "Diego", "score": 75, "seconds": 80},
    {"name": "Mei", "score": 88, "seconds": 101},
]

def by_seconds(player):
    """Sort key: how long the player took."""
    return player["seconds"]
```

```python
def print_leaderboard(players):
    """Print players ranked by score, highest first."""
    print("=== Quiz Leaderboard ===")
    ranked = sorted(players, key=by_score, reverse=True)
    for rank, player in enumerate(ranked, start=1):
        name = player["name"]
        score = player["score"]
        print(f"{rank}. {name:<8}{score:>4}")
```

```python
def print_summary(players):
    """Print the average score and the fastest player."""
    scores = []
    for player in players:
        scores.append(player["score"])
    fastest = min(players, key=by_seconds)
    print(f"Average score: {sum(scores) / len(scores):.1f}")
    print(f"Fastest: {fastest['name']} ({fastest['seconds']} s)")
    print(f"Gap between 1st and last: {max(scores) - min(scores)}")

print_leaderboard(players)
print_summary(players)
```

`by_score` works here too because these dicts also have a `"score"` key. A key function only cares that each item
gives it what it needs.
