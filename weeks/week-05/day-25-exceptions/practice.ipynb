- prints `There must be at least one student.` and returns `None` when `students` is 0.

Expected output (when finished):

````python
def safe_average(total_text, students_text):
    """Return total / students, or None (with a message) if the input is bad."""
    total = int(total_text)
    students = int(students_text)
    return total / students


print(safe_average("450", "6") == 75.0)
# print(safe_average("450", "0") == None)
# print(safe_average("four hundred", "6") == None)
````
