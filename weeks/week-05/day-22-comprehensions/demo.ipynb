## 1. Warm-up: the build-a-list loop

Since Day 12 we have built new lists like this. Spot the three parts: **start empty → loop → append**.

````python
raw_names = ["  asha ", "BILAL", "sara  ", "mEI"]

clean_names = []
for name in raw_names:
    clean_names.append(name.strip().title())

print(clean_names)
````

## 2. The list comprehension

The same result in one line:

````python
clean_names = [name.strip().title() for name in raw_names]
print(clean_names)
````

How the loop maps onto the comprehension:

```
clean_names = []                        clean_names = [ name.strip().title()  for name in raw_names ]
for name in raw_names:          ──►                     └─ what to keep ─┘     └─ the loop line ─┘
    clean_names.append(name.strip().title())
```

Read it aloud: *"name-strip-title, **for each** name **in** raw_names"*. The square brackets say "this makes a list".

🤔 **Predict** each list before running.

````python
print([n * 2 for n in [1, 2, 3]])
print([len(word) for word in ["hi", "hello", "hey"]])
print([letter.upper() for letter in "abc"])
````

A classic: Celsius → Fahrenheit for a week of temperatures.

````python
celsius = [18, 22.5, 30, -4]
fahrenheit = [c * 9 / 5 + 32 for c in celsius]
print(fahrenheit)
````

Very common in real programs: turning text into numbers (remember yesterday's `sum(["12", "30", "8"])` error).

````python
text = "12,7,30"
numbers = [int(part) for part in text.split(",")]
print(numbers, "total:", sum(numbers))
````

## 3. Filtering with `if`

Add `if condition` at the **end** to keep only some items.

````python
marks = [78, -5, 91, 105, 64, 0, 88]

valid_marks = [mark for mark in marks if mark >= 0 and mark <= 100]
print(valid_marks)
````

```
[ mark          for mark in marks      if mark >= 0 and mark <= 100 ]
  └ keep this ┘  └── the loop ──┘      └──── only when this is True ───┘
```

🤔 **Predict:**

````python
fruits = ["apple", "kiwi", "fig", "banana"]
print([fruit for fruit in fruits if len(fruit) > 3])
````

Transform **and** filter together: keep only the entries that are digits, and convert them.

````python
entries = ["12", "abc", "7", "", "30"]
quantities = [int(entry) for entry in entries if entry.isdigit()]
print(quantities)
````

Yesterday we built a list of `True`/`False` with four lines before calling `all()`. Now:

````python
print("Everyone passed:", all([mark >= 50 for mark in valid_marks]))
print("Anyone above 90:", any([mark > 90 for mark in valid_marks]))
````

## 4. Dict comprehensions

Curly brackets and `key: value` make a **dictionary**.

````python
names = ["Asha", "Bilal", "Mei"]
name_lengths = {name: len(name) for name in names}
print(name_lengths)
````

With `zip()` from Day 21, two lists become one dictionary:

````python
scores = [88, 75, 93]
score_book = {name: score for name, score in zip(names, scores)}
print(score_book)
````

Looping over `.items()` (Day 12) lets us filter or change a dictionary.

````python
prices = {"apple": 0.5, "laptop": 899.0, "pen": 1.2, "headphones": 59.99}

cheap = {item: price for item, price in prices.items() if price < 10}
on_sale = {item: round(price * 0.9, 2) for item, price in prices.items()}
print(cheap)
print(on_sale)
````

🤔 **Predict:** how many entries will this dictionary have?

````python
print({word: len(word) for word in ["hi", "hello", "hi"]})
````

Keys are unique, so the second `"hi"` just overwrites the first.

## 5. When a normal loop is clearer

A comprehension is a tool for **one job**: build a new list/dict from another collection.
This one works, but nobody can read it in one breath:

````python
raw_products = ["  usb-cable ", "", "LAPTOP-STAND", "123", " mouse-pad"]

clean = [p.strip().lower().replace("-", " ").title() for p in raw_products if p.strip() != "" and not p.strip().isdigit()]
print(clean)
````

**Fix 1:** move the complicated parts into small named functions (Week 4 skills). The comprehension becomes a sentence.

````python
def clean_product(text):
    """Tidy one product name: trim spaces, dashes to spaces, Title Case."""
    return text.strip().lower().replace("-", " ").title()


def is_product_name(text):
    """A product name is not blank and not just a number."""
    text = text.strip()
    return text != "" and not text.isdigit()


clean = [clean_product(p) for p in raw_products if is_product_name(p)]
print(clean)
````

**Fix 2:** decisions with `if/elif/else` belong in a loop **or** in a function, never squeezed inline.

````python
def letter_grade(mark):
    """Return A, B or C for a mark from 0 to 100."""
    if mark >= 85:
        return "A"
    elif mark >= 70:
        return "B"
    else:
        return "C"


print([letter_grade(mark) for mark in valid_marks])
````

`print()` returns `None`, so we built a useless list of `None`s. **Comprehensions build values; loops do actions**
(printing, saving, asking for input, `break`).

| Use a comprehension when… | Use a normal loop when… |
|---------------------------|-------------------------|
| you build a new list/dict from another collection | you print, save or ask for input |
| one transform, at most one `if` | you need `elif`/`else`, `break` or several steps |
| it fits on one line you can read aloud | you update something else (a total, a counter dict) |

## 6. Common mistakes (these cells fail **on purpose**)

````python
# Mistake 1: the filter if goes at the END, not before for
[mark if mark >= 50 for mark in marks]
````

````python
# Mistake 2: a dict comprehension needs key: value (a colon, not a comma)
{name, len(name) for name in names}
````

````python
# Mistake 3: the loop variable is `s`, but the expression uses `student`
students = ["omar", "fatima"]
[student.title() for s in students]
````

| Error message | Usual cause |
|---------------|-------------|
| `expected 'else' after 'if' expression` | filter `if` placed before `for` |
| `did you forget parentheses around the comprehension target?` | `,` instead of `:` in a dict comprehension |
| `NameError` | the expression uses a different name from the loop variable |

## 7. Real-world example: cleaning sign-up data

A sign-up form gave us messy text. Four short, readable one-liners turn it into usable data.

````python
raw_signups = ["  asha khan", "BILAL AHMED ", "", "sara ali", "   ", "mei chen"]
raw_marks = ["78", "abc", "91", "", "105", "64"]
temps_c = [12.5, 30, -2, 21]
````

````python
names = [s.strip().title() for s in raw_signups if s.strip() != ""]
marks = [int(m) for m in raw_marks if m.isdigit()]
valid = [m for m in marks if m <= 100]
temps_f = [round(c * 9 / 5 + 32, 1) for c in temps_c]

print(names)
print(valid)
print(temps_f)
````

````python
usernames = {name: name.lower().replace(" ", ".") for name in names}
print(usernames)

exam = {"Asha Khan": 78, "Bilal Ahmed": 49, "Sara Ali": 91, "Mei Chen": 64}
passed = [name for name, mark in exam.items() if mark >= 50]
print("Passed:", passed)
````

Notice we filtered `marks` in **two steps** (`isdigit`, then `<= 100`) instead of one long line. Short steps with
good names beat one clever line.

## Recap

1. `[expression for item in collection]` builds a new list: it replaces *start empty → loop → append*.
2. Add `if condition` **at the end** to keep only some items.
3. `{key: value for item in collection}` builds a dictionary; `.items()` and `zip()` work inside it.
4. Comprehensions **build values**. If you need printing, `elif`, `break` or many steps, use a loop or a helper function.
