````python
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages


class Shelf:
    def __init__(self, label):
        self.label = label
        self.books = []

    def add(self, book):
        self.books.append(book)

    def total_pages(self):
        return sum([book.pages for book in self.books])


novel = Book("Kindred", 264)
guide = Book("Python Basics", 150)
home = Shelf("Home")
school = Shelf("School")
home.add(novel)
home.add(guide)
school.add(guide)
guide.pages = 200
print(home.total_pages())                   # A
print(school.total_pages())                 # B
school.books[0].title = "Python Basics (2nd ed.)"
print(guide.title)                          # C
print(len(home.books), len(school.books))   # D
````

## Exercise 2: A `Course` of `Student`s

The `Student` class from Day 33 is given (run the cell). Write a class `Course`:

| Method | What it does |
|--------|--------------|
| `__init__(title)` | stores the title and an empty list `students` |
| `enrol(student)` | raises `ValueError("<id> is already enrolled.")` if a student with the same `student_id` is already in the course; otherwise adds the student |
| `find(student_id)` | returns the `Student` with that id, or `None` |
| `class_average()` | the average of the students' **averages** (ask each student!), or `0.0` for an empty course |
| `top_student()` | returns the `Student` **object** with the best average, or `None` for an empty course |
| `roster()` | **returns** a multi-line string: a title line, then one line per student (use `str(student)`) |

Then uncomment and run the check cell. Expected output:

```
Python Basics (3 students)
Asha (S001): 3 marks, average 82.3
Bilal (S002): 2 marks, average 67.0
Mei (S003): 4 marks, average 88.5
Class average: 79.3
Top student: Mei
Sorry: S002 is already enrolled.
None
```

````python
class Student:
    def __init__(self, name, student_id):
        self.name = name
        self.student_id = student_id
        self.marks = []

    def add_mark(self, mark):
        if mark < 0 or mark > 100:
            raise ValueError(f"A mark must be between 0 and 100, got {mark}.")
        self.marks.append(mark)

    def average(self):
        if len(self.marks) == 0:
            return 0.0
        return sum(self.marks) / len(self.marks)

    def __str__(self):
        return f"{self.name} ({self.student_id}): {len(self.marks)} marks, average {self.average():.1f}"


def make_student(name, student_id, marks):
    """Helper: create a Student and add all the marks."""
    student = Student(name, student_id)
    for mark in marks:
        student.add_mark(mark)
    return student
````
