My prediction:
- A: …

````python
class Locker:
    def __init__(self, number, owner="nobody"):
        self.number = number
        self.owner = owner
        self.items = []


first = Locker(12)
second = Locker(13, "Zara")
third = second
third.owner = "Ken"
first.items.append("umbrella")
print(first.number, first.owner, first.items)      # A
print(second.number, second.owner, second.items)   # B
````
