## 1. Class vs instance

A **class** is like a blank registration form: it decides which fields every form has.
An **instance** is one filled-in form. Each one has its own values.

You already use classes: `str` is a class, and every piece of text is an instance of it.

```python
first = "Asha"
second = "Bilal"
print(type(first), type(second))
```

Two different instances, one class. Today we write the class ourselves.

## 2. Our first class

```python
class Student:
    def __init__(self, name, student_id):
        self.name = name
        self.student_id = student_id
```

| Part | Meaning |
|------|---------|
| `class Student:` | the design of a new kind of object (class names use `CapitalWords`) |
| `__init__` | the **constructor**: runs automatically to set up each new object |
| `self` | **the specific object being set up right now** (the form being filled in) |
| `self.name = name` | store the value on *this* object: an **attribute** |

Running the cell above only *defined* the class. No student exists yet. Now we make one (an **instance**):

```python
asha = Student("Asha", "S001")
print(asha.name)
print(asha.student_id)
```

What happened on `Student("Asha", "S001")`:

1. Python created a new, empty `Student` object.
2. Python called `__init__` with `self` = that new object, `name="Asha"`, `student_id="S001"`.
3. `__init__` attached the attributes `name` and `student_id` to it.
4. The finished object came back and got the name tag `asha`.

We pass **two** arguments; `__init__` has **three** parameters. Python fills in `self` for us.

## 3. Many independent instances

```python
bilal = Student("Bilal", "S002")
print(asha.name, bilal.name)
```

🤔 **Predict:** we change Bilal's name. Does Asha's name change?

```python
bilal.name = "Bilal Ahmed"
print(asha.name)
print(bilal.name)
```

Each instance has its **own** attributes, like two separate filled-in forms.

```python
print(type(asha))
```

Our class is a type now, just like `str`. (`__main__` means "defined in this notebook".)

🤔 **Predict:** this time we do *not* call `Student(...)`. What happens to `asha`?

```python
same = asha
same.name = "Asha K."
print(asha.name)
```

`same = asha` attached a second name tag to the **same** object (Day 07). Only calling `Student(...)` makes a new
object.

## 4. Attributes in practice

Attributes do not all have to come from parameters: some start with a fixed value, some are calculated.

```python
class Student:
    def __init__(self, name, student_id, year=1):
        self.name = name
        self.student_id = student_id
        self.year = year              # default parameter value
        self.marks = []               # every student starts with an empty list
        self.email = f"{name.lower()}.{student_id.lower()}@school.edu"   # calculated


sara = Student("Sara", "S003")
omar = Student("Omar", "S004", year=2)
print(sara.year, sara.marks, sara.email)
print(omar.year, omar.marks, omar.email)
```

Attributes can be read, changed and (for lists) appended to from outside:

```python
sara.year = 2
sara.marks.append(78)
sara.marks.append(91)
print(sara.year, sara.marks)
print(omar.marks)
```

🤔 **Predict:** why is Omar's list still empty? (Each call to `__init__` ran `self.marks = []` and made a **new** list.)

## 5. Printing objects and lists of objects

```python
print(sara)
```

Not very friendly: the type and a memory address. Tomorrow we fix that with `__str__`. For now, use the attributes:

```python
def describe(student):
    return f"{student.name} ({student.student_id}), year {student.year}, marks {student.marks}"


print(describe(sara))
```

Notice: `describe` only makes sense for students. Tomorrow it moves **inside** the class.

A list of instances works like any list (Day 12):

```python
class_list = [Student("Mei", "S005"), Student("Diego", "S006"), Student("Fatima", "S007")]
class_list[0].marks.append(88)
class_list[0].marks.append(72)
class_list[1].marks.append(65)
class_list[2].marks.append(94)
class_list[2].marks.append(90)

for student in class_list:
    average = sum(student.marks) / len(student.marks)
    print(f"{student.name:<8}{average:>6.1f}")
```

## 6. Common mistakes (these cells fail on purpose)

Read the **last line** of each error.

```python
# Mistake 1: forgetting self in __init__
class Book:
    def __init__(title, author):
        self.title = title
        self.author = author


Book("Kindred", "Octavia Butler")
```

"takes 2 positional arguments but 3 were given": Python always passes the new object as the first argument,
so the count includes `self`.

```python
# Mistake 2: missing an argument
asha = Student("Asha")
```

```python
# Mistake 3: forgetting "self." (the value is only a local variable)
class Book:
    def __init__(self, title, author):
        title = title
        author = author


kindred = Book("Kindred", "Octavia Butler")
print(kindred.title)
```

```python
# Mistake 4: a typo in __init__ (one underscore each side), so Python never runs it
class Book:
    def _init_(self, title, author):
        self.title = title
        self.author = author


Book("Kindred", "Octavia Butler")
```

```python
# Mistake 5: a typo in an attribute name (read the suggestion at the end!)
print(sara.nmae)
```

## 7. Dict vs class (optional discussion)

| Day 31 dict | Day 32 class |
|-------------|--------------|
| `asha = {"name": "Asha", "student_id": "S001"}` | `asha = Student("Asha", "S001")` |
| `asha["name"]` | `asha.name` |
| shape defined only by habit | shape defined once, in `__init__` |
| typo key → silently a new key | typo while *reading* → `AttributeError` with a suggestion |
| `type(asha)` is `dict` | `type(asha)` is `Student` |

Honest note: `sara.nmae = "x"` (a typo while **setting**) still silently creates a new attribute. Tomorrow's methods

## Recap

1. A **class** is a template; an **instance** is one object made from it, with its own attributes.
2. `__init__` runs automatically when you call the class. `self` is the specific object being set up.
3. `self.name = name` stores an attribute; `asha.name` reads it.
4. Errors to recognise: missing `self`, wrong number of arguments, missing `self.`, misspelt `__init__`.
