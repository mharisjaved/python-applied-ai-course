## 1. Why NumPy?

The average of some quiz scores, the way we did it until now:

```python
scores = [72, 85, 58, 91, 64, 77]

total = 0
for score in scores:
    total += score
print(total / len(scores))
```

The same with NumPy. By convention we import it as `np`.

```python
import numpy as np

scores_array = np.array(scores)
print(scores_array.mean())
```

## 2. Arrays: `shape` and `dtype`

An **array** is like a list of numbers, but every item has the **same type** and maths works on all items at once.

```python
quiz = np.array([72, 85, 58, 91, 64, 77])
print(quiz)
print(type(quiz))
print("shape:", quiz.shape)    # (6,) = one dimension with 6 items
print("dtype:", quiz.dtype)    # the type shared by every item
```

🤔 **Predict:** one number has a decimal point. What will the `dtype` be?

```python
heights = np.array([1.62, 1.75, 2, 1.80])
print(heights)
print(heights.dtype)
```

One float makes the **whole array** float: every item must have the same type. (`int64`/`float64` just means
"integer/decimal number stored in 64 bits".)

## 3. Vectorised arithmetic: maths on every item

"Vectorised" means one operation is applied to **every item**, with no loop written by us.

```python
print(quiz + 5)         # 5 bonus points for everyone
print(quiz / 100 * 20)  # convert to a mark out of 20
```

🤔 **Predict both lines before running.** Lists and arrays react very differently to `* 2`.

```python
print([1, 2, 3] * 2)
print(np.array([1, 2, 3]) * 2)
```

Two arrays of the **same length** combine item by item: first with first, second with second…

```python
exam = np.array([60, 75, 50, 88])
coursework = np.array([70, 65, 80, 90])
final = exam * 0.7 + coursework * 0.3
print(final)
```

```python
# Mistake: arrays of different lengths cannot be paired item by item
np.array([60, 75, 50]) + np.array([70, 65, 80, 90])
```

### How much faster?

```python
import time   # time.perf_counter() reads a stopwatch, in seconds

numbers = list(range(1_000_000))   # 1_000_000 = one million; the _ only makes it easier to read
numbers_array = np.array(numbers)

start = time.perf_counter()
doubled_list = []
for n in numbers:
    doubled_list.append(n * 2)
loop_seconds = time.perf_counter() - start

start = time.perf_counter()
doubled_array = numbers_array * 2
numpy_seconds = time.perf_counter() - start

print(f"loop:  {loop_seconds:.4f} s")
print(f"numpy: {numpy_seconds:.4f} s")
```

NumPy is many times faster (often 10× or more): it runs the loop for us in fast compiled code. The exact numbers
depend on the computer, and the gap grows with the size of the data.

## 4. Summary statistics

```python
quiz = np.array([72, 85, 58, 91, 64, 77])
print("mean:  ", quiz.mean())
print("median:", np.median(quiz))
print("min:   ", quiz.min(), "| max:", quiz.max())
print("std:   ", round(quiz.std(), 1))
```

- `median` = the middle value after sorting (here, the average of the two middle values 72 and 77).
  It is a function `np.median(...)`, not a method.
- `.argmax()` gives the **position** of the largest value: handy to find *who* scored highest.

```python
names = ["Asha", "Bilal", "Sara", "Omar", "Mei", "Diego"]
best = quiz.argmax()
print(best, names[best], quiz[best])
```

### Mean vs median

🤔 **Predict:** Ken was absent and got 0. Which will move more, the mean or the median?

```python
with_ken = np.array([72, 85, 58, 91, 64, 77, 0])
print("mean:  ", round(with_ken.mean(), 1))
print("median:", np.median(with_ken))
```

One extreme value drags the mean down a lot; the median barely moves. When a few values are extreme, report the median too.

### Standard deviation = "typical distance from the average"

🤔 **Predict:** both classes have the same mean (70). Which one has the bigger `std`?

```python
class_a = np.array([68, 70, 71, 69, 72])
class_b = np.array([40, 95, 60, 85, 70])
print("A mean", class_a.mean(), "std", round(class_a.std(), 1))
print("B mean", class_b.mean(), "std", round(class_b.std(), 1))
```

In class A everyone is close to 70 (std ≈ 1.4). In class B scores are spread out (std ≈ 19.2). Same average,
very different classes. The mean alone hides that.

## 5. Boolean masks: filtering without loops

A comparison on an array gives an array of `True`/`False`, one per item. We call it a **mask**.

```python
quiz = np.array([72, 85, 58, 91, 64, 77])
passed = quiz >= 65
print(passed)
```

```python
print(quiz[passed])          # keep only the items where the mask is True
print(passed.sum())          # True counts as 1, False as 0 → how many passed
print(passed.mean())         # share of students who passed
```

🤔 **Predict:** how many scores are between 60 and 80 (inclusive)?
Combine masks with `&` (and) and `|` (or). **Each condition needs its own parentheses.**

```python
middle = (quiz >= 60) & (quiz <= 80)
print(quiz[middle], middle.sum())
```

```python
# Mistake 1: Python's `and` cannot combine two whole arrays
(quiz >= 60) and (quiz <= 80)
```

```python
# Mistake 2: no parentheses. `&` is done first (60 & quiz), then the comparison fails
quiz >= 60 & quiz <= 80
```

A mask can also **change** the selected items. Add 10 bonus points, but never above 100:

```python
curved = quiz + 10
curved[curved > 100] = 100
print(curved)
```

## 6. 2-D arrays: rows and columns

A list of lists becomes a **table**. Here: 4 students (rows) × 3 tests (columns).

```python
grades = np.array([
    [70, 82, 64],   # Asha
    [55, 61, 70],   # Bilal
    [90, 88, 95],   # Fatima
    [62, 45, 58],   # Ken
])
print(grades.shape)   # (rows, columns)
```

Index with `[row, column]`. A `:` means "all of them".

```python
print(grades[0])        # row 0: all of Asha's tests
print(grades[2, 1])     # row 2, column 1: Fatima's second test
print(grades[:, 0])     # every row, column 0: the first test for everyone
```

🤔 **Predict:** `axis=0` goes *down* the rows (one result per column); `axis=1` goes *across* (one result per row).
Which line gives each **test's** average and which gives each **student's** average?

```python
print(grades.mean(axis=0))
print(grades.mean(axis=1).round(1))
```

## 7. Random numbers with a seed

Simulated data is useful for testing ideas. A **seed** makes the "random" numbers repeatable.

```python
rng = np.random.default_rng(42)
print(rng.integers(0, 101, size=5))        # 5 whole numbers from 0 to 100
print(rng.normal(70, 10, size=5).round(1)) # 5 numbers around 70, typical distance 10
```

🤔 **Predict:** if we create the generator again with the same seed, do we get the same numbers?

```python
rng = np.random.default_rng(42)
print(rng.integers(0, 101, size=5))
```

Same seed → same numbers → same results for everyone in the room. Next week scikit-learn calls this `random_state`.

## 8. Real-world example: vectorised grade statistics

Load the cleaned student file with `csv` (as on Day 41), then turn each column into an array.

```python
import csv
from pathlib import Path

DATA_DIR = Path("../../../datasets")
with open(DATA_DIR / "students.csv", encoding="utf-8", newline="") as f:
    rows = list(csv.DictReader(f))

final = np.array([float(row["final_score"]) for row in rows])
hours = np.array([float(row["study_hours"]) for row in rows])
program = np.array([row["program"] for row in rows])
print(final.shape, hours.shape, program.shape)
```

```python
print(f"Students: {len(final)}")
print(f"Mean {final.mean():.1f} | median {np.median(final):.1f} | std {final.std():.1f}")
print(f"Lowest {final.min()} | highest {final.max()}")
print(f"Pass rate: {(final >= 50).mean() * 100:.1f}%")
```

🤔 **Predict:** do students who study 10+ hours a week score higher on average? By how much?

```python
many_hours = hours >= 10
print(f"10+ hours:  {final[many_hours].mean():.1f}  ({many_hours.sum()} students)")
print(f"under 10 h: {final[~many_hours].mean():.1f}  ({(~many_hours).sum()} students)")
```

`~` flips a mask: `True` becomes `False` and the other way round ("not").

Masks work on text arrays too:

```python
for name in ["science", "commerce", "arts"]:
    in_program = program == name
    print(f"{name:<9} {in_program.sum():>3} students, mean final {final[in_program].mean():.1f}")
```

Finally, a 2-D table of features, **rows = students, columns = features**. This is exactly the shape scikit-learn
expects next week.

```python
features = np.array([[float(row["study_hours"]), float(row["attendance_pct"]), float(row["sleep_hours"])]
                     for row in rows])
print(features.shape)
print(features[:3])                  # first three students
print(features.mean(axis=0).round(2))  # average of each feature
```

## Recap

1. `np.array(list)` → array. Every item has the same `dtype`; `shape` gives its size.
2. Arithmetic is **vectorised**: `scores + 5` changes every item, no loop.
3. `.mean()`, `np.median()`, `.std()`, `.min()`, `.max()`, `.argmax()`. Std = typical distance from the mean.
4. Masks: `arr[arr >= 50]`, combine with `&` `|` and parentheses; `mask.sum()` counts, `mask.mean()` gives a share.
5. 2-D: `[row, column]`, `[:, col]`, `axis=0` per column, `axis=1` per row. Seeds make randomness repeatable.
