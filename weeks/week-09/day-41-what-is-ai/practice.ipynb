Remember: every value from `csv.DictReader` is **text**. Convert with `float()` or `int()` before comparing numbers.

````python
# Setup: run me first
import csv
from pathlib import Path

DATA_DIR = Path("../../../datasets")

with open(DATA_DIR / "students.csv", encoding="utf-8", newline="") as f:
    students = list(csv.DictReader(f))


def predict_pass(student, threshold):
    """Predict 1 (pass) if the student studies at least `threshold` hours a week, else 0 (fail)."""
    if float(student["study_hours"]) >= threshold:
        return 1
    return 0


def accuracy(students, threshold):
    """Share of students (0 to 1) whose result the rule 'study_hours >= threshold' predicts correctly."""
    correct = 0
    for student in students:
        if predict_pass(student, threshold) == int(student["passed"]):
            correct += 1
    return correct / len(students)


print(len(students), "students loaded. Columns:", list(students[0].keys()))
````
