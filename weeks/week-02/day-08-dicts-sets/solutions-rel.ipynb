## Exercise 1: Predict the output

| Expression | Answer | Common wrong answer |
|------------|--------|---------------------|
| `menu["juice"]` | `3.0` | |
| `len(menu)` | `3` | `6` (counting keys and values) |
| `"cake" in menu` | `False` | |
| `3.0 in menu` | `False` | `True`: `in` checks **keys**, not values |
| `menu.get("cake")` | `None` | an error |
| `menu.get("cake", 0)` | `0` | |
| `sorted(menu)` | `['juice', 'samosa', 'tea']` | a sorted list of prices or pairs |
| `max(menu.values())` | `3.0` | |
| `len({"red", "blue", "red"})` | `2` | `3` |
| `sorted({3, 1, 2} & {2, 3, 4})` | `[2, 3]` | `[1, 2, 3, 4]` (that is the union) |

```python
menu = {"tea": 2.5, "samosa": 1.2, "juice": 3.0}
print(menu["juice"])
print(len(menu))
print("cake" in menu)
print(3.0 in menu)
print(menu.get("cake"))
print(menu.get("cake", 0))
print(sorted(menu))
print(max(menu.values()))
print(len({"red", "blue", "red"}))
print(sorted({3, 1, 2} & {2, 3, 4}))
```

## Exercise 2: Contact book

```python
contacts = {"Asha": "555-0142", "Bilal": "555-0178", "Sara": "555-0193"}

print("Sara:", contacts["Sara"])
contacts["Zara"] = "555-0101"
contacts["Bilal"] = "555-0222"
del contacts["Asha"]
print(contacts)
print("Omar:", contacts.get("Omar", "not found"))
print("Contacts:", len(contacts))
print("Names:", sorted(contacts))
```

Watch for:
- `contacts.append(...)` or `contacts.add(...)`: dictionaries add with `contacts[key] = value`.
- Students surprised that updating Bilal kept his **position**: updating changes the value, not the order.
- `sorted(contacts.keys())` is also correct; `sorted(contacts)` is shorter because sorting a dict sorts its keys.

## Exercise 4: Student record

```python
student = {
    "name": "Omar Farooq",
    "id": "S027",
    "marks": [64, 71, 59],
    "courses": ["Python", "Statistics"],
}

print(f"{student['name']} ({student['id']})")

student["marks"].append(80)
average = sum(student["marks"]) / len(student["marks"])
print(f"Average: {average:.1f} | Best: {max(student['marks'])}")

student["courses"].append("AI Basics")
print("Courses:", len(student["courses"]))

first_name = student["name"].split()[0]
print("First name:", first_name)

student["year"] = 2
print(student)
```

Common wrong answers:
- `student["marks"] = 80` replaces the whole list with one number. Ask: "what type should marks be?"
- `student.append(80)`: dictionaries have no `.append`; the **list inside** does.
- f-string quote clash: `f"{student["name"]}"` works in Python 3.12+, but older versions reject it. Teach the
  single-quote habit.
- `first_name, last_name = student["name"].split()` (unpacking) is a nice alternative for part 5.

```python
class_list = [
    {"name": "Fatima", "score": 79},
    {"name": "Ken", "score": 47},
    {"name": "Diego", "score": 88},
]

print("Second student:", class_list[1]["name"])
print("Last score:", class_list[-1]["score"])
class_list[1]["score"] = 74
print(class_list[1])
```

## Exercise 5: Course enrolments overlap

```python
python_signups = ["Asha", "Bilal", "Sara", "Omar", "Asha", "Ken", "Bilal"]
ai_signups = ["Sara", "Mei", "Omar", "Diego", "Mei", "Zara"]

python_students = set(python_signups)
ai_students = set(ai_signups)
print(f"Python: {len(python_students)} students | AI: {len(ai_students)} students")
print("Both:", sorted(python_students & ai_students))
print(f"At least one course: {len(python_students | ai_students)} students")
print("Python only:", sorted(python_students - ai_students))

ai_students.add("Fatima")
print("Fatima in AI?", "Fatima" in ai_students)
```

Watch for:
- `len(python_signups)` (7) instead of the set's length (5): the duplicates are still in the list.
- `ai_signups.add("Fatima")` → `AttributeError`: lists use `.append`, sets use `.add`.

## Exercise 6 (challenge): Who missed the quizzes?

```python
class_list = {"Asha", "Bilal", "Sara", "Omar", "Mei", "Diego", "Zara", "Ken"}
quiz_1 = ["Asha", "Sara", "Omar", "Mei", "Asha", "Zara"]
quiz_2 = ["Sara", "Bilal", "Mei", "Ken", "Mei", "Fatima"]

done_1 = set(quiz_1)
done_2 = set(quiz_2)
did_any = done_1 | done_2
did_both = done_1 & done_2

report = {
    "missed_quiz_1": sorted(class_list - done_1),
    "missed_both": sorted(class_list - did_any),
    "not_in_class": sorted(did_any - class_list),
    "exactly_one": sorted((did_any - did_both) & class_list),
}
print(report)
```

Reasoning:
- "Missed both" = everyone minus anyone who did at least one quiz.
- "Exactly one" = (did any) minus (did both). Without `& class_list`, Fatima appears in it; that is the most
  common wrong answer and a good conversation about checking data against a master list.
- Naming the in-between sets (`did_any`, `did_both`) makes the final dictionary readable. Encourage it.
- Python also has a `^` operator (symmetric difference) for "exactly one"; mention it only if a student finds it.
