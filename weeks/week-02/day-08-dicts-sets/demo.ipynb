## 1. A dictionary: look up by key

```python
phone_book = {
    "Asha": "555-0142", 
    "Bilal": "555-0178", 
    "Sara": "555-0193"
}
print(phone_book["Bilal"])
```

A **dictionary** (`dict`) stores **key → value** pairs inside curly brackets `{ }`.
Like a real dictionary (word → meaning) or a cloakroom (ticket number → coat): you hand over the key, you get the value.

| key | value |
|-----|-------|
| `"Asha"` | `"555-0142"` |
| `"Bilal"` | `"555-0178"` |
| `"Sara"` | `"555-0193"` |

```python
import json
print(json.dumps(phone_book, indent=2))

# print(phone_book)
print(len(phone_book), "contacts")
print(type(phone_book))
```

## 2. Missing keys: `KeyError`, `in` and `.get()`

🤔 **Predict:** what happens when we ask for someone who is not in the phone book?

```python
phone_book["Zara"]
```

`in` checks whether a **key** exists, before we ask for it:

```python
print("Zara" in phone_book)
print("Bilal" in phone_book)
```

🤔 **Predict:** `in` checks **keys**. So what does this give?

```python
print("555-0142" in phone_book)
```

`.get(key, default)` never crashes: it gives back the value, or the default if the key is missing.

```python
print(phone_book.get("Asha", "not saved"))
print(phone_book.get("Zara", "not saved"))
print(phone_book.get("Zara"))      # no default given: None ("nothing", Day 07)
```

## 3. Adding, updating and deleting

```python
import json
print(json.dumps(phone_book, indent=2))
```

```python
phone_book["Zara"] = "555-0101"    # a new key is added
print(json.dumps(phone_book, indent=2))
```

🤔 **Predict:** Asha has a new number. Will the dictionary have two `"Asha"` entries now?

```python
phone_book["Asha"] = "555-0199"
print(phone_book)
```

**Keys are unique.** Assigning to an existing key **replaces** its value. The same line adds or updates.

```python
del phone_book["Sara"]
print(phone_book)
```

Values can be numbers that we update, like stock in a shop:

```python
stock = {"apples": 12, "bananas": 30}
stock["apples"] += 24     # a delivery arrived
stock["bananas"] -= 5     # sold 5
print(stock)
```

## 4. Looking inside: `.keys()`, `.values()`, `.items()`

```python
prices = {"tea": 2.50, "samosa": 1.20, "juice": 3.00, "cake": 2.75}
print(prices.keys())
print(prices.values())
print(prices.items())
```

Our Day 06 tools work on these:

```python
print("Items on the menu:", len(prices))
print("Cheapest price:", min(prices.values()))
print(f"One of everything: {sum(prices.values()):.2f}")
print("Menu A-Z:", sorted(prices))      # sorting a dict sorts its keys
```

`list()` turns them into an ordinary list, just as `int()` turns text into a number. Each item is a
**(key, value) tuple**, so we can unpack it (Day 07).

```python
menu_pairs = list(prices.items())
print(menu_pairs)
item, price = menu_pairs[0]
print(f"{item}: {price:.2f}")
```

On Day 12 we will **loop** over `.items()` to visit every pair.

## 6. Lists of dicts and nested dicts

A **list of dictionaries** is how a table of rows looks in Python (we will meet it again with CSV files in Week 8).

```python
class_list = [
    {"name": "Asha", "score": 82},
    {"name": "Omar", "score": 67},
    {"name": "Mei", "score": 91},
]
print(class_list[1].get("name"), class_list[1].get("score"))
print(class_list[1]["name"], class_list[1]["score"])
```

🤔 **Predict** `class_list[2]["score"]` and `class_list[-1]["name"]`.

```python
print(class_list[2]["score"], class_list[-1]["name"])
```

A **nested dictionary** lets each contact hold several facts:

```python
contacts = {
    "Asha": {"phone": "555-0142", "city": "Lahore"},
    "Bilal": {"phone": "555-0178", "city": "Karachi"},
}
print(contacts["Bilal"]["city"])
contacts["Asha"]["city"] = "Islamabad"
print(contacts["Asha"])
```

## 7. Sets: unique values only

```python
signups = ["Asha", "Omar", "Asha", "Mei", "Omar", "Asha"]
unique_signups = set(signups)
print(unique_signups)
print(len(signups), "sign-ups from", len(unique_signups), "different people")
```

A **set** keeps each value **once** and has **no order** (your neighbour may see a different order). Because
there is no order, there are no positions. Use `sorted()` to get a tidy list back.

```python
print(sorted(unique_signups))
```

```python
tags = {"python", "ai", "data"}

tags.add("python")      # already there: nothing changes
tags.add("beginner")

print(tags)

print(len(tags), "ai" in tags)
```

## 8. Comparing groups: `|`, `&`, `-`

```python
python_class = {"Asha", "Bilal", "Sara", "Omar"}
ai_class = {"Sara", "Omar", "Mei", "Diego"}

print("In both courses:   ", sorted(python_class & ai_class))
print("In at least one:   ", sorted(python_class | ai_class))
print("Python but not AI: ", sorted(python_class - ai_class))
```

- `a | b` (**union**): in `a` **or** `b` (or both)
- `a & b` (**intersection**): in **both**
- `a - b` (**difference**): in `a` but **not** in `b`

🤔 **Predict:** `ai_class - python_class`.

```python
print(sorted(ai_class - python_class))
```

## 10. Which structure should I use?

| I need… | Use | Example |
|---------|-----|---------|
| an ordered collection that grows or gets sorted | `list` | shopping list, marks |
| a fixed record that should not change | `tuple` | `(year, month, day)` |
| to look things up by a name or ID | `dict` | phone book, student record |
| unique values, or overlaps between groups | `set` | who takes both courses |

```python
enrolments = {
    "python": ["Asha", "Bilal", "Sara", "Asha"],    # Asha signed up twice
    "ai": ["Sara", "Mei", "Diego"],
}
python_students = set(enrolments["python"])
ai_students = set(enrolments["ai"])
print("Python:", len(python_students), "| AI:", len(ai_students))
print("Taking both:", sorted(python_students & ai_students))
```

## Recap

1. A dict maps **keys → values**: `d[key]` looks up, `d[key] = value` adds or updates, `del d[key]` removes.
2. A missing key raises `KeyError`. Check with `key in d`, or use `d.get(key, default)`.
3. `.keys()`, `.values()`, `.items()` show the contents; `sum/min/max/sorted/list()` work on them.
4. Records: dicts can hold lists; lists can hold dicts. Read `a[...][...]` left to right.
5. A set keeps unique values with no order. `set(list)` removes duplicates; `| & -` compare groups.
