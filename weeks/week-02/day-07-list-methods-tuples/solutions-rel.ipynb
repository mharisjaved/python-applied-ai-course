## Exercise 1: Trace the list

**Snippet A**, step by step:

| Line | `tasks` | `done` |
|------|---------|--------|
| start | `['email', 'report', 'call']` | |
| `.append("lunch")` | `['email', 'report', 'call', 'lunch']` | |
| `.insert(0, "coffee")` | `['coffee', 'email', 'report', 'call', 'lunch']` | |
| `done = tasks.pop(2)` | `['coffee', 'email', 'call', 'lunch']` | `'report'` |
| `.remove("call")` | `['coffee', 'email', 'lunch']` | `'report'` |

Prints: `['coffee', 'email', 'lunch']`, `report`, `3`.
Most common wrong answer: `pop(2)` removing `'call'` (index counted on the list *before* the insert).

```python
tasks = ["email", "report", "call"]
tasks.append("lunch")
tasks.insert(0, "coffee")
done = tasks.pop(2)
tasks.remove("call")
print(tasks)
print(done)
print(len(tasks))
```

**Snippet B** prints `[1, 2, 5, 9]`, `None`, `[1, 2, 5, 9]`. `b` is `None` because `.sort()` changes the list
and gives back nothing. Students who predicted `b` = `[1, 2, 5, 9]` have met the trap; ask them for the rule.

```python
nums = [5, 2, 9, 1]
a = sorted(nums)
b = nums.sort()
print(a)
print(b)
print(nums)
```

## Exercise 2: Shopping-list operations

```python
shopping_text = "rice, lentils, onions, tea"

shopping = shopping_text.split(", ")
shopping.append("yogurt")
shopping.insert(0, "salt")
shopping.remove("lentils")
bought = shopping.pop(0)
shopping.sort()

print("Bought:", bought)
print(f"Still to buy ({len(shopping)}): " + ", ".join(shopping))
```

Common wrong answers:
- `.split(",")` (no space): items become `' lentils'` with a leading space, so `.remove("lentils")` fails with
  `ValueError`. Good teaching moment: print the list and look at the quotes.
- `bought = shopping.pop()` (no index) removes the **last** item, `'yogurt'`.
- `shopping = shopping.sort()` → `None`, then `len(None)` fails.

Alternative for the last line: `print(f"Still to buy ({len(shopping)}): {', '.join(shopping)}")`.

## Exercise 4: Tuples and unpacking

```python
appointment = ("Dr. Rahman", "Tuesday", "10:30", 2)   # doctor, day, time, room
cities = [("Lahore", 31.5), ("Karachi", 24.9), ("Islamabad", 33.7)]   # (city, latitude)

doctor, day, time, room = appointment
print(f"Your appointment with {doctor} is on {day} at {time} in room {room}.")

new_appointment = (doctor, "Thursday", time, room)
print(new_appointment)

print(cities[1][0])
name, lat = cities[-1]
print(f"{name} is at latitude {lat}")
```

Watch for:
- `print(cities[1])` prints the whole tuple `('Karachi', 24.9)`. They need a second index for the name.
- Students who write `appointment[1] = "Thursday"` get the tuple `TypeError`; ask why a tuple is a good choice
  for an appointment record (the parts belong together and should not be edited one by one).

## Exercise 5: Two names, one list

Part 1 prints `Monday: ['gym', 'laundry', 'dentist']`. Explanation: `tuesday_tasks = monday_tasks` puts a second
name tag on the **same** list, so appending through either name changes that one list.

```python
monday_tasks = ["gym", "laundry"]
tuesday_tasks = monday_tasks.copy()
tuesday_tasks.append("dentist")
print("Monday:", monday_tasks)
print("Tuesday:", tuesday_tasks)
```

Part 2: `sorted()` gives a new list, so the original order is kept. `marks.sort(reverse=True)` would destroy
the register order. That is the wrong answer to look for.

```python
marks = [70, 55, 90]

print("Ranked:", sorted(marks, reverse=True))
print("Register order:", marks)
```

Alternative: `ranked = sorted(marks)` then `ranked.reverse()`. Also fine, and it shows that `.reverse()` only
affects the copy.

## Exercise 6 (challenge): Exam leaderboard

```python
names = ["Asha", "Bilal", "Sara", "Omar", "Mei", "Diego", "Zara"]
scores = [81, 67, 94, 75, 94, 58, 88]

top = max(scores)
top_position = scores.index(top)
print(f"Top scorer: {names[top_position]} with {top}")
print("Students with the top score:", scores.count(top))

ranked = sorted(scores, reverse=True)
print("Top 3 scores:", ranked[:3])

low_to_high = sorted(scores)
print("Median score:", low_to_high[len(low_to_high) // 2])

lowest_position = scores.index(min(scores))
left = names.pop(lowest_position)
scores.pop(lowest_position)
print(f"{left} left. Students remaining: {len(names)}")
```

Reasoning to discuss:
- `.index()` returns the **first** match, which is why Sara (not Mei) is the top scorer.
- Median: 7 sorted scores, the middle is index `7 // 2 = 3`. For an even count you would average the two middle
  values; mention it only if asked.
- Part 5 must remove the **same position** from both lists, otherwise the lists stop lining up. Some students
  use `names.remove("Diego")`, which types a name from the list; ask them to make it work for any data.
- This exercise shows the pain of parallel lists. Tomorrow's dictionaries keep each name and score together.
