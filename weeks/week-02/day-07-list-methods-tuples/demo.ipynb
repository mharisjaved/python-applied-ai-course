## 1. Adding items: `.append()` and `.insert()`

```python
cart = ["milk", "eggs", "bread"]
cart.append("tea")
print(cart)
```

🤔 **Predict:** where will `"rice"` end up?

```python
cart.insert(1, "rice")
print(cart)
```

- `.append(item)` adds **one** item at the **end**.
- `.insert(index, item)` puts the item **at that position** and shifts the rest to the right.

A very common pattern: start with an empty list and fill it.

```python
visitors = []
visitors.append("Omar")
visitors.append("Mei")
print(visitors, len(visitors))
```

## 2. Removing items: `.remove()` and `.pop()`

```python
cart = ["milk", "rice", "eggs", "bread", "tea", "eggs"]
cart.remove("eggs")
print(cart)
```

🤔 There were **two** `"eggs"`. How many are left? `.remove()` deletes the **first** match only.

`.pop()` removes an item **by position** and **gives it back**, so you can keep using it.

```python
queue = ["Ken", "Zara", "Diego"]
served = queue.pop(0)          # the first person in the queue
print("Now serving:", served)
print("Still waiting:", queue)
```

```python
last = queue.pop()             # no index: the last item
print(last, queue)
```

| Method | Removes | Gives back |
|--------|---------|------------|
| `.remove(value)` | the first item equal to `value` | nothing |
| `.pop()` | the last item | that item |
| `.pop(index)` | the item at `index` | that item |

## 3. Finding: `.count()` and `.index()`

```python
attendance = ["P", "P", "A", "P", "L", "P", "A"]   # 7 sessions: Present, Absent, Late
print("Absences:", attendance.count("A"))
print("First absence in session", attendance.index("A") + 1)
```

🤔 Why `+ 1`? Because `.index()` counts from 0, but people count sessions from 1.

Now we can answer yesterday's question: **who** scored the highest mark?

```python
students = ["Asha", "Bilal", "Sara", "Omar", "Mei", "Diego"]
marks = [72, 85, 68, 95, 58, 77]

top_position = marks.index(max(marks))
print(students[top_position], "scored", max(marks))
```

The two lists are **parallel**: position 3 in `marks` belongs to position 3 in `students`.

## 4. Sorting: `.sort()` versus `sorted()`

```python
marks = [72, 85, 68, 95, 58, 77]
marks.sort()
print(marks)
marks.sort(reverse=True)
print(marks)
```

`reverse=True` is an **option** we pass to the method: "sort from biggest to smallest".
`.reverse()` just flips the current order, without sorting:

```python
arrivals = ["Zara", "Asha", "zainab", "Mei"]
arrivals.reverse()
print(arrivals)
```

`sorted()` is a built-in tool (like `len`) that **gives back a new sorted list** and leaves the original alone.

```python
prices = [4.50, 1.25, 3.00]
cheapest_first = sorted(prices)
print("Sorted copy (cheapest copy):", cheapest_first)
print("Original:   ", prices)
print("Dearest first:", sorted(prices, reverse=True))   # same option as .sort()
```

🤔 **Predict:** what does this print?

```python
marks = [72, 85, 68]
result = marks.sort()
print(result)
print(marks)
```

`None` is Python's value for **"nothing here"** (more about it on Day 17).
`.sort()` changes the list **in place** and gives back `None`.

| Changes the list itself, gives back `None` | Gives back a new value |
|------|------|
| `.append` `.insert` `.remove` `.sort` `.reverse` | `sorted()` `.count()` `.index()` slices; `.pop()` changes the list **and** gives back the item |

**Rule:** never write `my_list = my_list.sort()`.

Sorting text puts capital letters **before** lowercase letters:

```python
print(sorted(["mei", "Zara", "asha", "Omar"]))
```

## 5. Text ↔ list: `.split()` and `.join()`

```python
sentence = "learning python one step at a time"
words = sentence.split()
print(words)
print(len(words), "words")
```

```python
line = "rice,lentils,onions,tea"
items = line.split(",")
print(items)
```

🤔 **Predict** both lines.

```python
print(" | ".join(items))
print(", ".join(sorted(items)))
```

- `text.split()` cuts at spaces; `text.split(",")` cuts at commas. The result is a **list of strings**.
- `"glue".join(list_of_strings)` sticks the items together into **one string**, with the glue in between.

## 6. Tuples: lists that cannot change

A **tuple** uses round brackets. Indexing, slicing, `len` and `in` work as for lists.

```python
birthday = (2004, 9, 17)        # year, month, day
location = (31.52, 74.36)       # latitude, longitude
print(birthday[0], len(birthday), type(birthday))
```

```python
# birthday[0] = 2005
birthday
```

Use a tuple for a **fixed record** where changing one piece would be a mistake: a date, a map coordinate, a
(name, student ID) pair. Use a list for a collection that grows, shrinks or gets sorted.

## 7. Unpacking: several names in one line

```python
year, month, day = birthday
# print(day, month, year)

print(year)
print(month)
print(day)
```

The number of names on the left must match the number of items on the right. Unpacking works on lists too:

```python
first, last = "Mei Lin".split()
print(f"{last.upper()}, {first}")
```

## 8. Two name tags, one list

Day 02 recap: moving one name tag does not move the other.

```python
a = 7
b = a
a = 100
print(a, b)
```

🤔 Now the same idea with a list, but this time we **change** the list instead of moving a tag. Predict `cart`.

```python
# deep copy, shallow copy.

cart = ["milk", "eggs"] 
shared = cart
shared.append("tea")
print(cart)
```

`shared = cart` does **not** copy the list. It sticks a **second name tag on the same list**. `.append` changes
that one list, so both names see the change. (Day 02 was different: there we *moved* a tag to a new value.)

To get an independent list, make a copy:

```python
cart = ["milk", "eggs"]
backup = cart.copy()
cart.append("tea")
print("cart:  ", cart)
print("backup:", backup)
```

## 10. Real-world example: shopping-list operations

```python
raw = "milk, eggs, bread, rice"
shopping = raw.split(", ")
shopping.append("tea")
shopping.insert(0, "onions")    # urgent, so put it at the top
shopping.remove("bread")        # found some at home
bought = shopping.pop(0)        # bought the first item
shopping.sort()

print("Bought:", bought)
print(f"Still to buy ({len(shopping)}): " + ", ".join(shopping))
```

## Recap

1. Add with `.append(x)` / `.insert(i, x)`; remove with `.remove(x)` (by value) or `.pop(i)` (by position, gives it back).
2. `.count(x)` and `.index(x)` answer "how many?" and "where?".
3. `.sort()` changes the list and returns `None`; `sorted()` returns a new list.
4. `.split()` turns text into a list; `"glue".join(list)` turns a list into text.
5. Tuples are fixed records; unpack them with `a, b = pair`.
6. `b = a` puts two name tags on **one** list. Use `.copy()` for an independent list.
