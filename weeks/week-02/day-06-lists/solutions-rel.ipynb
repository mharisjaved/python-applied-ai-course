## Exercise 1: Predict the output

| Expression | Answer | Common wrong answer |
|------------|--------|---------------------|
| `days[1]` | `'Tue'` | `'Mon'` (counting from 1) |
| `days[-2]` | `'Sat'` | |
| `days[2:5]` | `['Wed', 'Thu', 'Fri']` | including `'Sat'` (stop is excluded) |
| `days[:3]` | `['Mon', 'Tue', 'Wed']` | |
| `days[5:]` | `['Sat', 'Sun']` | |
| `days[::2]` | `['Mon', 'Wed', 'Fri', 'Sun']` | |
| `len(days)` | `7` | `6` (confusing length with last index) |
| `"Sun" in days` | `True` | |
| `"sun" in days` | `False` | `True` (case matters) |

Point out: an index gives a **string**, a slice gives a **list** (look for the square brackets in the output).

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
print(days[1])
print(days[-2])
print(days[2:5])
print(days[:3])
print(days[5:])
print(days[::2])
print(len(days))
print("Sun" in days)
print("sun" in days)
```

## Exercise 2: Class marks report

```python
marks = [66, 81, 74, 59, 93, 88, 70, 62]

average = sum(marks) / len(marks)
print("Number of marks:", len(marks))
print("Total:", sum(marks))
print("Highest:", max(marks))
print("Lowest:", min(marks))
print(f"Average: {average:.1f}")
```

Common wrong answers:
- `sum(marks) / 8`: works today, breaks when a mark is added. Ask "what if a ninth student joins?"
- `print("Average:", round(average, 1))` is a fine alternative (Day 03).
- `average = sum(marks) / len` (forgot the brackets and the list) → `TypeError`.

## Exercise 4: Update the playlist

```python
playlist = ["Monsoon Morning", "Night Bus Home", "Chai at Midnight", "City of Kites", "Slow Tide"]

print("First:", playlist[0], "| Last:", playlist[-1])

playlist[2] = "Paper Boats"   # the third song is at index 2
print(playlist)

print("Up next:", playlist[1:4])
print("Slow Tide" in playlist)
print("Rainy Window" not in playlist)
print(f"The playlist has {len(playlist)} songs.")
```

Watch for:
- `playlist[3] = "Paper Boats"`: "third" means index **2**.
- `playlist[1:3]` for "positions 1, 2 and 3": the stop index is excluded, so it must be `4`.
- An f-string alternative for line 1: `print(f"First: {playlist[0]} | Last: {playlist[-1]}")`.

## Exercise 5: A week of temperatures

```python
temps = [28.4, 30.1, 33.6, 35.0, 31.2, 27.9, 29.5]   # Mon, Tue, Wed, Thu, Fri, Sat, Sun

weekend = temps[5:]
print(f"Hottest: {max(temps)} °C")
print(f"Coldest: {min(temps)} °C")
print(f"Range: {max(temps) - min(temps):.1f} °C")
print(f"Average: {sum(temps) / len(temps):.1f} °C")
print("Weekend:", weekend)
print(f"Weekend average: {sum(weekend) / len(weekend):.1f} °C")
print("Was it exactly 35.0 on any day?", 35.0 in temps)

temps[2] = 32.6   # Wednesday is index 2
print(f"Corrected average: {sum(temps) / len(temps):.1f} °C")
```

Discussion points:
- Without `:.1f` the range prints as `7.100000000000001`. Floats are stored approximately; formatting hides it.
  Mention it briefly. Do **not** go into binary floating point.
- `temps[-2:]` is an equally good weekend slice, and it keeps working if more days are added at the start.
- Weekend average: `(temps[5] + temps[6]) / 2` is correct but less general than `sum(weekend) / len(weekend)`.

## Exercise 6 (challenge): Gradebook with nested lists

```python
names = ["Asha", "Bilal", "Sara"]
scores = [
    [78, 85, 90],   # Asha
    [64, 70, 58],   # Bilal
    [91, 88, 95],   # Sara
]

print(f"{names[1]}, test 2: {scores[1][1]}")
print(f"{names[2]}'s best: {max(scores[2])}")
print(f"{names[0]}'s average: {sum(scores[0]) / len(scores[0]):.1f}")

scores[1][2] = 72
print(f"{names[1]} after the re-sit: {scores[1]}")

test_1_total = scores[0][0] + scores[1][0] + scores[2][0]
print(f"Class average on test 1: {test_1_total / len(scores):.1f}")
```

- First index = student (row), second index = test (column). Students often swap them: `scores[2][1]` vs `scores[1][2]`.
- Part 5 cannot use `sum()` directly because the test-1 scores are spread over three lists. With 30 students the
  sum would need 30 terms, which is exactly the problem from section 1 of the demo. **Loops (Day 11)** solve it;
  use this as the hook for Week 3.
