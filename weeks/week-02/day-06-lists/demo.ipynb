## 1. The problem: one variable per value

```python
mark_1 = 72
mark_2 = 85
mark_3 = 64
mark_4 = 90
mark_5 = 58
average = (mark_1 + mark_2 + mark_3 + mark_4 + mark_5) / 5
print(average)
```

Fine for 5 students. For 30 students we need 30 names and a very long formula, and every new student means
editing the formula. We need **one name for the whole group**.

## 2. A list: many values under one name

A **list** keeps several values **in order**, inside square brackets `[ ]`, separated by commas.

```python
marks = [72, 85, 64, 90, 58]
print(marks)
print(type(marks))
```

Picture a row of numbered lockers. The name tag `marks` is attached to the **whole row**, and every locker has a
position number (an **index**) starting at **0**.

| index | 0 | 1 | 2 | 3 | 4 |
|-------|---|---|---|---|---|
| value | 72 | 85 | 64 | 90 | 58 |
| negative index | -5 | -4 | -3 | -2 | -1 |

```python
students = ["Aisha", "Bilal", "Sara", "Omar", "Mei"]   # a list of strings
empty_basket = []                                     # an empty list, to be filled later
print(students)
print(empty_basket)
```

## 3. Indexing and slicing: the same rules as strings

🤔 Predict each line before running.

```python
print(students[0])
print(students[-1])
print(students[1:3])
```

- An **index** gives one **item**: `students[0]` is the string `"Asha"`.
- A **slice** gives a **new list**: `students[1:3]` holds positions 1 and 2 (the stop position is not included).

```python
print(students[:2])     # the first two
print(students[-2:])    # the last two
print(students[::-1])   # a reversed copy
```

```python
playlist = ["Monsoon Morning", "Night Bus Home", "Chai at Midnight", "City of Kites", "Slow Tide"]
print("Now playing:", playlist[0])
print("Up next:", playlist[1:4])
print("Last song:", playlist[-1])
```

🤔 **Predict:** `students` has 5 items. What happens with `students[5]`?

```python
students[5]
```

Valid positions are `0` to `len - 1`, so the last one here is `4`. To get the last item, use `students[-1]`.

## 4. Built-in tools: `len`, `sum`, `min`, `max`

```python
temperatures = [31.5, 33.0, 29.8, 35.2, 34.1, 30.6, 32.4]   # one week, in °C
print("Days:", len(temperatures))
print("Hottest:", max(temperatures))
print("Coldest:", min(temperatures))
print(f"Total: {sum(temperatures):.1f}")
```

🤔 **Predict:** there is no built-in `average()`. How can we calculate the average from the tools above?

```python
average_temp = sum(temperatures) / len(temperatures)
print(f"Average: {average_temp:.1f} °C")
print(f"Range:   {max(temperatures) - min(temperatures):.1f} °C")
```

`sum(...) / len(...)` keeps working when the list grows. That is the payoff from section 1.

`min` and `max` also work on strings. They use alphabetical order:

```python
students = ["Aisha", "Abisha", "Bilal", "Sara", "Omar", "Mei", "Zainab"] 

print(min(students), max(students))
```

## 5. Changing an item

Lists can be changed after they are created. Assign a new value to a position:

```python
marks = [72, 85, 64, 90, 58]
marks[2] = 68          # Sara's paper was re-marked
print(marks)
```

🤔 **Predict:** Omar (index 3) gets 5 bonus marks. What does the list look like after this cell?

```python
marks[3] = marks[3] + 5
print(marks)
```

Right side first (Day 02): read `marks[3]` (90), add 5, store 95 back in position 3. `marks[3] += 5` does the same.

Strings are different: they **cannot** be changed in place (Day 04). Lists can.

```python
word = "Python"
word[0] = "J"
```

## 6. Is it in the list? `in` and `not in`

`in` asks a yes/no question, so the answer is a `bool`.

```python
students = ["Aisha", "Abisha", "Bilal", "Sara", "Omar", "Mei", "Zainab"] 

print("Omar" in students)
print("Zara" in students)
print("Zara" not in students)
```

🤔 **Predict** both results.

```python
print("omar" in students)
print("Om" in students)
```

- On a list, `in` looks for an **exact, whole item**. Capital letters matter, and `"Om"` is not an item.
- On a string, `in` looks for a **piece of text**:

```python
print("Om" in "Omar")
```

## 7. Preview: a list inside a list

A list can hold other lists, like the rows of a table. Here is a classroom seating plan:

```python
seating = [
    ["Asha", "Bilal", "Sara"],    # row 0
    ["Omar", "Mei", "Diego"],     # row 1
]
print(seating[1])
print(seating[1][0]) # unpacking
```

🤔 **Predict** `len(seating)` and `seating[0][-1]`.

```python
print(len(seating))
print(seating[0][-1])
```

Read the indexes left to right: first choose the row, then the seat in that row. Data tables in Week 9 work the same way.

## 8. Common mistakes (these cells fail **on purpose**)

```python
# Mistake 1: round brackets instead of square brackets
marks(0)
```

```python
# Mistake 2: the numbers were stored as text
marks_text = int["72", "85", "64"]
sum(marks_text)
```

```python
# Mistake 3: max() needs at least one item
max([])
```

A sneaky one with **no error message**: a missing comma glues two strings into one item.

```python
team = ["Fatima" "Ken", "Zara"]
print(team, len(team))
```

## 9. Real-world example: a class-marks report

```python
students = ["Asha", "Bilal", "Sara", "Omar", "Mei", "Diego"]
marks = [72, 85, 68, 95, 58, 77]

print(f"Students: {len(students)}")
print(f"Average:  {sum(marks) / len(marks):.1f}")
print(f"Highest:  {max(marks)}")
print(f"Lowest:   {min(marks)}")
print(f"First in the register: {students[0]} with {marks[0]}")
print("Is Zara in this class?", "Zara" in students)
```

## Recap

1. A list stores many values in order under one name: `marks = [72, 85, 64]`.
2. Index and slice exactly as with strings. The first index is `0`, the last is `-1`.
3. `len`, `sum`, `min`, `max` summarise a list. Average = `sum(x) / len(x)`.
4. Lists can be changed: `marks[2] = 70`. Strings cannot.
5. `in` / `not in` check whether an item is in the list.
