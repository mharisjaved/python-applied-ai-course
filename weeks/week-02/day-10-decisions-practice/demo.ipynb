## 1. The method: spec → tests → code → run the tests

1. **Read the spec.** Underline the **inputs**, the **output** and every **rule**.
2. **Write a test table before coding:** one row per rule, plus the **boundaries** (exactly 100, just below 100…)
   and at least one **invalid** input. Work out the expected output by hand.
3. **Write the code:** inputs at the top, decisions in the middle, output at the end.
4. **Run every row.** Change the inputs, re-run the cell, mark ✅ or ❌. After a fix, re-run **all** rows.

## 2. Worked example: shipping cost

> **Spec.** An online shop charges shipping by zone: `local` 3.00, `national` 6.50, `international` 18.00.
> Orders of **100.00 or more** ship free, but only to local and national addresses.
> Parcels **heavier than 5 kg** pay an extra 4.00, even when shipping is otherwise free.
> An unknown zone prints `Unknown zone: <zone>`. Otherwise print `Shipping: 7.50` (two decimals).

**Inputs:** `zone`, `order_total`, `weight_kg` · **Output:** one line of text.

**Test table** (written *before* the code):

| # | zone | order_total | weight_kg | Expected | Why this row |
|---|------|-------------|-----------|----------|--------------|
| 1 | local | 40.00 | 2.0 | Shipping: 3.00 | base price |
| 2 | international | 40.00 | 2.0 | Shipping: 18.00 | base price |
| 3 | national | 100.00 | 2.0 | Shipping: 0.00 | **boundary**: exactly 100 is free |
| 4 | national | 99.99 | 2.0 | Shipping: 6.50 | just below the boundary |
| 5 | international | 150.00 | 2.0 | Shipping: 18.00 | never free |
| 6 | local | 40.00 | 5.0 | Shipping: 3.00 | **boundary**: 5 kg is not *heavier than* 5 |
| 7 | national | 120.00 | 8.0 | Shipping: 4.00 | free **and** heavy |
| 8 | mars | 40.00 | 2.0 | Unknown zone: mars | invalid input |

A first attempt, run with **row 3**:

```python
rates = {"local": 3.00, "national": 6.50, "international": 18.00}

zone = "national"
order_total = 100.00
weight_kg = 2.0

if zone not in rates:
    print(f"Unknown zone: {zone}")
else:
    cost = rates[zone]
    if order_total > 100 and zone != "international":
        cost = 0.0
    if weight_kg > 5:
        cost += 4.00
    print(f"Shipping: {cost:.2f}")
```

🤔 Expected `Shipping: 0.00`. Which row of the spec did the code get wrong, and which character fixes it?

Row 3 ❌. The spec says "100.00 **or more**", so the comparison must be `>=`. Rows 1, 2 and 4 would all have
passed, which is why the **boundary rows** are in the table.

```python
rates = {"local": 3.00, "national": 6.50, "international": 18.00}

zone = "national"
order_total = 100.00
weight_kg = 2.0

if zone not in rates:
    print(f"Unknown zone: {zone}")
else:
    cost = rates[zone]
    if order_total >= 100 and zone != "international":
        cost = 0.0
    if weight_kg > 5:
        cost += 4.00
    print(f"Shipping: {cost:.2f}")
```

Row 3 ✅. Now change the three input lines to each other row, re-run, and tick the table.

**Design question:** why are "free shipping" and "heavy parcel" two separate `if`s and not an `if` / `elif`?

- `elif` means **choose one** of several options (a grade is A *or* B *or* C).
- Separate `if`s mean **check each rule on its own**: a parcel can be free **and** heavy (row 7).

## 3. The same program with real input

Users type with capitals and spaces, so clean the text and convert the numbers first.

```python
rates = {"local": 3.00, "national": 6.50, "international": 18.00}

zone = input("Zone (local/national/international): ").strip().lower()
order_total = float(input("Order total: "))
weight_kg = float(input("Weight in kg: "))

if zone not in rates:
    print(f"Unknown zone: {zone}")
else:
    cost = rates[zone]
    if order_total >= 100 and zone != "international":
        cost = 0.0
    if weight_kg > 5:
        cost += 4.00
    print(f"Shipping: {cost:.2f}")
```

🤔 Which row of the test table did that run match? (Row 7: `"  National "` became `"national"`.)

## 4. A good test table has…

- ✅ one row for **every branch** (every `if`, `elif` and `else` block runs at least once);
- ✅ both sides of every **boundary** (`100` and `99.99`; `5` and `5.5`);
- ✅ at least one **invalid** input (unknown zone, negative age, empty text);
- ✅ **messy text** if the input is typed by a person (`"  National "`, `"NATIONAL"`);
- ✅ the expected output written down **before** running the code.

## 5. Common mistakes in decision problems (this cell fails **on purpose**)

```python
# A variable set in only some branches
minutes = 20

if minutes > 30:
    fee = 2.0
    
print(f"Fee: {fee:.2f}")
```

`fee` is only created when the `if` is true. Fix: give it a starting value **before** the `if` (`fee = 0.0`), or
set it in **every** branch (add an `else`).

Other mistakes that give **no error**, only wrong answers, and which only a test table catches:
- `>` where the spec says "or more" (`>=`), or the other way round;
- an `elif` where two rules can both apply (or separate `if`s where only one should);
- `day == "Sat" or "Sun"`;
- comparing text that was not cleaned with `.strip().lower()`.

## Recap

1. Spec → test table → code → run every row. Write the expected output **before** you run the code.
2. Test every branch, both sides of every boundary, and invalid or messy input.
3. `elif` = choose one; separate `if`s = independent rules.
4. Make sure every variable you print is set in every path through the code.
