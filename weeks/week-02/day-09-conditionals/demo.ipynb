## 1. Comparisons give `True` or `False`

```python
score = 72
print(score >= 50)
print(score == 100)
print(score != 100)

# score == 100
```

| Operator | Means | Example (`score = 72`) |
|----------|-------|------------------------|
| `==` | equal to | `score == 72` → `True` |
| `!=` | not equal to | `score != 72` → `False` |
| `<` / `>` | less / greater than | `score > 80` → `False` |
| `<=` / `>=` | less / greater than or equal to | `score >= 72` → `True` |

`=` **attaches a name** (Day 02). `==` **asks a question**: "are these equal?"

## 4. `if` / `elif` / `else`: many paths

Grade boundaries: A ≥ 85 · B ≥ 70 · C ≥ 55 · D ≥ 40 · otherwise F.

```python
score = 78
if score >= 85:
    grade = "A"
elif score >= 70:
    grade = "B"
elif score >= 55:
    grade = "C"
elif score >= 40:
    grade = "D"
else:
    grade = "F"
print(f"Score {score} → grade {grade}")
```

Python checks the conditions **from top to bottom** and runs **only the first** branch that is `True`. The rest
are skipped, even if they are also true (78 is also `>= 55`).

🤔 **Predict:** same boundaries, different order. What grade does 78 get now?

```python
score = 78

if score >= 40:
    grade = "D"
elif score >= 55:
    grade = "C"
elif score >= 70:
    grade = "B"
elif score >= 85:
    grade = "A"
else:
    grade = "F"
print(grade)
```

No error, wrong answer. With `>=`, put the **strictest** (highest) condition first.

## 5. Combining conditions: `and`, `or`, `not`

| Operator | `True` when… | Example |
|----------|--------------|---------|
| `and` | **both** sides are true | `total >= 50 and city == "Lahore"` |
| `or` | **at least one** side is true | `age < 18 or is_student` |
| `not` | flips `True` ↔ `False` | `not is_member` |

```python
age = 20
is_student = True
print(age < 18 or is_student)
print(age < 18 and is_student)
print(not is_student)
```

```python
order_total = 64.0
city = "Lahore"
is_member = True

if order_total >= 50 and city == "Lahore":
    print("Free delivery")
elif is_member:
    print("Members' delivery fee: 2.00")
else:
    print("Delivery fee: 5.00")
```

Checking a range: both lines below mean "between 0 and 100". Python allows the shorter, chained form.

```python
score = 105
print(score >= 0 and score <= 100)
print(0 <= score <= 100)
```

```python
day = "Mon"
if day == "Sat" or "Sun":
    print("Weekend!")
else:
    print("Weekday")
```

Python reads it as `(day == "Sat") or ("Sun")`. The right side is just the text `"Sun"`, which counts as true

```python
if day == "Sat" or day == "Sun":
    print("Weekend!")
else:
    print("Weekday")
    
# days = ["sat", "sun", "mon"]

# "sunday" in days
```

## 6. Nested `if`: a decision inside a decision

```python
catalogue = {"Python Basics": 2, "Data Stories": 0}
title = "Data Stories"

if title in catalogue:
    if catalogue[title] > 0:
        catalogue[title] -= 1
        print(f"Borrowed '{title}'. Copies left: {catalogue[title]}")
    else:
        print(f"'{title}' is on loan. Join the waiting list.")
else:
    print(f"We don't stock '{title}'.")
```

The **indentation** shows which `else` belongs to which `if`. Checking `title in catalogue` first means the
inner lookup `catalogue[title]` can never raise a `KeyError`.

## 7. Truthiness: empty means `False`

`bool()` converts a value to `True` or `False`. 🤔 Predict each one.

```python
print(bool(0), bool(""), bool([]), bool({}))
print(bool(7), bool("hi"), bool([0]))
print(bool(" "), bool("0"))
```

**Rule:** zero and **empty** values (`0`, `""`, `[]`, `{}`, `set()`) count as `False`. Everything else counts as
`True`, even `" "` (one space) and `"0"` (a non-empty string). So `if` can test "is there anything in it?" directly:

```python
# name = "   "
# if name.strip():
#     print(f"Hello, {name.strip()}!")
# else:
#     print("Please type your name.")

if " ":
    print(True)
else:
    print(False)
```

```python
cart = []
if cart:
    print("Items in cart:", len(cart))
else:
    print("Your cart is empty.")
```

## 9. Real-world example: a grade calculator

```python
score = float(input("Enter the exam score (0-100): "))

if 0 <= score <= 100:
    if score >= 85:
        grade = "A"
    elif score >= 70:
        grade = "B"
    elif score >= 55:
        grade = "C"
    elif score >= 40:
        grade = "D"
    else:
        grade = "F"
    print(f"Score {score:.1f} → grade {grade}")
else:
    print("Invalid score: it must be between 0 and 100.")
```
