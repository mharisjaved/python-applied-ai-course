## Exercise 1: Predict, then check

**Part A**

| Expression | Answer | Why / common mistake |
|------------|--------|----------------------|
| `x > 5` | `True` | |
| `x == 7.0` | `True` | an int and a float with the same value are equal |
| `x != 7` | `False` | |
| `name == "mei"` | `False` | capitals matter |
| `name.lower() == "mei"` | `True` | |
| `x > 5 and x < 7` | `False` | `x < 7` is false, and `and` needs both |
| `x < 5 or name == "Mei"` | `True` | `or` needs only one |
| `not items` | `True` | an empty list counts as `False`; `not` flips it |
| `"5" == 5` | `False` | text is never equal to a number (no error, just `False`) |
| `3 <= x <= 10` | `True` | chained comparison |

```python
x = 7
name = "Mei"
items = []
print(x > 5)
print(x == 7.0)
print(x != 7)
print(name == "mei")
print(name.lower() == "mei")
print(x > 5 and x < 7)
print(x < 5 or name == "Mei")
print(not items)
print("5" == 5)
print(3 <= x <= 10)
```

## Exercise 2: Two paths with `if` / `else`

**a)** `14` → even, `9` → odd, `0` → even (0 % 2 is 0).

```python
number = 14

if number % 2 == 0:
    print(number, "is even")
else:
    print(number, "is odd")
```

**b)** With `balance = 200.0` it prints `Purchase approved. New balance: 50.01`.

```python
balance = 120.0
price = 149.99

if balance >= price:
    balance = balance - price
    message = f"Purchase approved. New balance: {balance:.2f}"
else:
    shortfall = price - balance
    message = f"Not enough money: you need {shortfall:.2f} more."
print(message)

print(message == "Not enough money: you need 29.99 more.")
```

Watch for:
- `balance > price` instead of `>=`: a customer with exactly enough money is refused. Ask "what if balance is 149.99?"
- printing `price - balance` without `:.2f` → `29.990000000000009`.

## Exercise 4: Grade calculator

```python
score = 78   # change this value to test every row of the table

if 0 <= score <= 100:
    if score >= 85:
        grade = "A"
    elif score >= 70:
        grade = "B"
    elif score >= 55:
        grade = "C"
    elif score >= 40:
        grade = "D"
    else:
        grade = "F"
    print(f"Score {score} → grade {grade}")
    if grade == "A":
        print("Well done!")
    elif grade == "F":
        print("Please book a tutoring session.")
else:
    print("Invalid score")

print(grade == "B")
```

Expected for the test table: 78 B · 85 A · 84 B · 40 D · 39 F · 100 A · 101 Invalid · -5 Invalid.

Common wrong answers:
- `score > 85` for an A: 85 gets a B. The test table catches it; this is why boundaries are in the table.
- Validation forgotten: 101 gets an A. Ask "what should a typo like 101 do?"
- Writing each band with both ends (`score >= 70 and score <= 84`): correct, but longer. Point out that the
  `elif` order already guarantees the upper limit.

Alternative layout (validation as the first branch of one chain):

```python
score = 101

if score < 0 or score > 100:
    print("Invalid score")
elif score >= 85:
    print(f"Score {score} → grade A")
elif score >= 70:
    print(f"Score {score} → grade B")
elif score >= 55:
    print(f"Score {score} → grade C")
elif score >= 40:
    print(f"Score {score} → grade D")
else:
    print(f"Score {score} → grade F")
```

It works, but repeats the `print` five times. Storing the grade in a variable and printing once (first version) is
easier to change. Good discussion for strong students.

## Exercise 5: Scholarship check

```python
average = 86.5
attendance_pct = 88
has_unpaid_fees = False

is_eligible = not has_unpaid_fees and (average >= 95 or (average >= 80 and attendance_pct >= 90))
print(is_eligible)
# print(is_eligible == False)
```

The brackets matter. Without them, `and` is evaluated before `or`, which changes the meaning. Tell students:
"when you mix `and` and `or`, always add brackets".

```python
if has_unpaid_fees:
    message = "Not eligible: unpaid fees"
elif average >= 95:
    message = "Eligible"
elif average < 80:
    message = "Not eligible: average below 80"
elif attendance_pct < 90:
    message = "Not eligible: attendance below 90%"
else:
    message = "Eligible"

print(message)
print(message == "Not eligible: attendance below 90%")
```

Reasoning: by the time Python reaches `elif attendance_pct < 90`, it already knows there are no unpaid fees and
the average is between 80 and 95, so only attendance is left to check. Each `elif` can rely on everything above it
being false. That is the key idea of the chain.

Most common wrong answer: checking attendance before the `average >= 95` branch, so the 96.0 / 70 row fails.

## Exercise 6 (challenge): Library help desk

```python
catalogue = {"python basics": 2, "data stories": 0, "ai for everyone": 1}
waiting_list = []
request = "  Data Stories "   # pretend the user typed this

title = request.strip().lower()
if not title:
    print("Please type a title.")
elif title not in catalogue:
    print(f"Sorry, we don't stock '{title.title()}'.")
elif catalogue[title] == 0:
    waiting_list.append(title)
    print(f"'{title.title()}' is on loan. You are number {len(waiting_list)} on the waiting list.")
else:
    catalogue[title] -= 1
    print(f"Enjoy '{title.title()}'! Copies left: {catalogue[title]}")

print(waiting_list == ["data stories"])
```

Expected for the other tests:
- `"PYTHON basics"` → `Enjoy 'Python Basics'! Copies left: 1`
- `"Cooking 101"` → `Sorry, we don't stock 'Cooking 101'.`
- `"   "` → `Please type a title.` (`"".strip()` is empty, so `not title` is `True`)

Discuss: why must the empty check come first? (*An empty title is also "not in the catalogue", so the wrong
message would appear.*) Nested `if` blocks are an equally valid answer; the flat `elif` chain is easier to read.
