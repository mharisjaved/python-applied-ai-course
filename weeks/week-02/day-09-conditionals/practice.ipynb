## Exercise 6 (challenge): Library help desk

A user types the title of a book they want. Write code that:

1. cleans the request (remove spaces at both ends, make it lowercase) so it matches the catalogue keys;
2. if the cleaned request is **empty**, prints `Please type a title.` (use truthiness, not `len`);
3. if the book is not in the catalogue, prints `Sorry, we don't stock 'Cooking 101'.` (title in Title Case);
4. if there are no copies left, adds the title to `waiting_list` and prints
   `'Data Stories' is on loan. You are number 1 on the waiting list.`;
5. otherwise takes one copy out of the catalogue and prints `Enjoy 'Python Basics'! Copies left: 1`.

Test with: `"  Data Stories "`, `"PYTHON basics"`, `"Cooking 101"`, `"   "`.

````python
catalogue = {"python basics": 2, "data stories": 0, "ai for everyone": 1}
waiting_list = []
request = "  python basics "   # pretend the user typed this

# 1. ---------------------------
request = request.strip().lower()

# 2. ---------------------------
if not request:
    print("Please type a title.")

# 3. ---------------------------
# request = "No Longer Human"
if request not in catalogue:
    print(f"Sorry, we don't stock {request.title()}")

# 4. ---------------------------
if catalogue.get(request) == 0:
    waiting_list.append(request)
    print(f"{request} is on loan, you are number 1 on the waiting list.")
else:
    catalogue[request] -= 1
    print(f"Enjoy {request}!")
    print(f"copies left {catalogue[request]}")


# if request:
#     pass
# else:
#     print("please type a title!")

# your code here
````
