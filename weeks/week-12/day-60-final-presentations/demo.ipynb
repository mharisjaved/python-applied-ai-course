## 2. Peer feedback: two stars and a wish

Each listener gives every presenter three quick scores (1–5) and one sentence: *two things that were good (stars)
and one thing to improve (a wish)*. Here is how to summarise the scores, using a small class (Week 7) and pandas
(Week 9). The feedback below is invented for the demo.

```python
class Presentation:
    """Peer feedback for one presenter."""

    def __init__(self, presenter, project):
        self.presenter = presenter
        self.project = project
        self.scores = []  # list of dicts: {"clear": 1-5, "works": 1-5, "responsible": 1-5}

    def add_feedback(self, clear, works, responsible):
        for value in [clear, works, responsible]:
            if value < 1 or value > 5:
                raise ValueError(f"Scores must be 1-5, got {value}")
        self.scores.append({"clear": clear, "works": works, "responsible": responsible})

    def __str__(self):
        return f"{self.presenter}: {self.project} ({len(self.scores)} feedback forms)"


mei = Presentation("Mei", "Campus FAQ assistant")
mei.add_feedback(5, 4, 4)
mei.add_feedback(4, 5, 3)
mei.add_feedback(5, 5, 4)
print(mei)
```

```python
import pandas as pd

feedback = pd.DataFrame(mei.scores)
feedback.mean().round(1)
```
