## 2. How to explain a function

Use four questions for any function. Here is one from Day 39 as a model.

```python
def parse_amount(text):
    """Turn text such as ' 12.50 ' into a positive float. Raises ValueError with a friendly message."""
    text = text.strip()
    if text == "":
        raise ValueError("Please type an amount.")
    try:
        amount = float(text)
    except ValueError:
        raise ValueError(f"'{text}' is not a number.")
    if amount <= 0:
        raise ValueError("The amount must be more than 0.")
    return round(amount, 2)
```

1. **What goes in?** "Text typed by the user, for example `' 12.50 '`."
2. **What comes out?** "A float rounded to 2 decimals, `12.5`. It returns, it does not print."
3. **What can go wrong, and what happens then?** "Empty text, a word, zero or a negative number. Each one raises
   `ValueError` with a message the menu shows to the user before asking again."
4. **How do you know it works?** Show the test:

```python
def test_parse_amount():
    assert parse_amount(" 12.50 ") == 12.5
    for bad in ["", "twelve", "0", "-5"]:
        try:
            parse_amount(bad)
            assert False, f"{bad} was not rejected"
        except ValueError:
            pass


test_parse_amount()
print("test passed")
```

`assert False, "message"` always fails with that message, so the test fails if `parse_amount(bad)` does *not*
raise. In a pytest file, `with pytest.raises(ValueError):` does the same job more neatly.

## 3. Warm-up: the Python phase in six predictions

🤔 Predict each output before running. Each cell revisits a different part of Weeks 1–8.

```python
# Weeks 2-3: dicts and loops
scores = {"Asha": 82, "Ken": 74}
scores["Mei"] = 91
print(len(scores), max(scores.values()))
```

```python
# Week 5: comprehensions
print([n * 2 for n in range(6) if n % 2 == 1])
```

```python
# Week 4: functions returning two values
def low_high(numbers):
    """Return the smallest and largest number."""
    return min(numbers), max(numbers)


low, high = low_high([4, 9, 1])
print(high - low)
```

```python
# Week 5: exceptions
try:
    hours = int("4.5")
except ValueError:
    hours = 0
print(hours)
```

```python
# Week 7: classes and __str__
class Plant:
    """A plant that needs watering every few days."""

    def __init__(self, name, every_days):
        self.name = name
        self.every_days = every_days

    def __str__(self):
        return f"{self.name} (water every {self.every_days} days)"


print(Plant("Basil", 2))
```

```python
# Week 8: JSON round trip
import json

saved = json.dumps({"plant": "Basil", "watered": ["2026-12-01"], "indoor": True})
print(saved)
print(json.loads(saved)["watered"][0])
```
