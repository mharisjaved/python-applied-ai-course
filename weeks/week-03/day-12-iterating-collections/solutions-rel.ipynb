## Exercise 1: Predict the loops

| | Output | Common wrong answer |
|-|--------|---------------------|
| a | `h`, `i`, `!` on separate lines | forgetting that `!` is a character too |
| b | `tea`, `cake`, then `tea 4`, `cake 10` | `2` and `5` for the first loop (it gives keys, not values) |
| c | `0 Zara`, `1 Ken` | `1 Zara`, `2 Ken` (no `start=1`, so counting starts at 0) |
| d | `4` | `2` or `5` (count carefully: Mi**ss**i**ss**ippi) |

```python
for ch in "hi!":
    print(ch)

prices = {"tea": 2, "cake": 5}
for item in prices:
    print(item)
for item, price in prices.items():
    print(item, price * 2)

for i, name in enumerate(["Zara", "Ken"]):
    print(i, name)

n = 0
for ch in "Mississippi":
    if ch == "s":
        n += 1
print(n)
```

## Exercise 2: Numbered shopping list

```python
shopping = ["rice", "eggs", "tomatoes", "yoghurt"]

print("Shopping list")
for number, item in enumerate(shopping, start=1):
    print(f"{number}. {item.title()}")
print(len(shopping), "items to buy")
```

Common mistakes: forgetting `start=1` (list starts at `0.`), or keeping a separate counter variable.
A counter works (it is Day 11's pattern) and is worth accepting, then show `enumerate` as the cleaner tool:

```python
number = 1
for item in shopping:
    print(f"{number}. {item.title()}")
    number += 1
```

## Exercise 3: Fix the counters

| | Mistake | Error | Fix |
|-|---------|-------|-----|
| a | looping over the dict gives keys only | `ValueError: too many values to unpack` | `stock.items()` |
| b | adding to a count that does not exist | `KeyError: 'red'` | `.get(colour, 0) + 1` |
| c | `upper_words = upper_words.append(...)` | `AttributeError` on the second pass | call `.append` without assigning |

```python
stock = {"pens": 40, "notebooks": 15, "rulers": 0}
for item, qty in stock.items():
    print(item, qty)
```

```python
colours = ["red", "blue", "red", "green", "blue", "red"]
colour_counts = {}
for colour in colours:
    colour_counts[colour] = colour_counts.get(colour, 0) + 1
print(colour_counts)
print(colour_counts == {"red": 3, "blue": 2, "green": 1})
```

```python
words = ["sun", "moon", "star"]
upper_words = []
for word in words:
    upper_words.append(word.upper())
print(upper_words)
```

## Exercise 4: Class report

One loop can do all three jobs. The `if/elif/else` chain must test the highest bar first (Day 09).

```python
students = [
    {"name": "Asha", "mark": 81},
    {"name": "Bilal", "mark": 45},
    {"name": "Sara", "mark": 67},
    {"name": "Omar", "mark": 92},
    {"name": "Mei", "mark": 58},
    {"name": "Diego", "mark": 39},
    {"name": "Fatima", "mark": 74},
]
grade_counts = {"A": 0, "B": 0, "C": 0, "F": 0}

passed_names = []
top_name = ""
top_mark = -1
for student in students:
    mark = student["mark"]

    if mark >= 50:
        passed_names.append(student["name"])

    if mark >= 80:
        grade_counts["A"] += 1
    elif mark >= 60:
        grade_counts["B"] += 1
    elif mark >= 50:
        grade_counts["C"] += 1
    else:
        grade_counts["F"] += 1

    if mark > top_mark:
        top_mark = mark
        top_name = student["name"]

print("Passed:", passed_names)
print("Grades:", grade_counts)
print(f"Top student: {top_name} with {top_mark}")
```

```python
print(passed_names == ["Asha", "Sara", "Omar", "Mei", "Fatima"])
print(grade_counts == {"A": 2, "B": 2, "C": 1, "F": 2})
```

Most common wrong answers:
- `if mark >= 50 … elif mark >= 60 …`: every passing mark stops at the first test, so everyone is C.
- `for name, mark in students:` (treating the records like a dict). Each record is one dict, so use
  `student["mark"]`.
- `top_mark = students[0]` (a whole dict) and then `mark > top_mark`, which gives a `TypeError`.

Slower students may use three separate loops; that is fine. Point out that one loop does it in a single pass.

## Exercise 5: Word counter

```python
feedback = "The class was fun and the exercises were fun but the homework was long and the room was cold"

word_counts = {}
for word in feedback.lower().split():
    word_counts[word] = word_counts.get(word, 0) + 1
print(word_counts)
print("Different words:", len(word_counts))

top_word = ""
top_count = 0
for word, count in word_counts.items():
    if count > top_count:
        top_count = count
        top_word = word
print(f"Most common word: '{top_word}' ({top_count} times)")
```

```python
print(word_counts["the"] == 4, len(word_counts) == 12)
```

Common mistakes: forgetting `.lower()` (then `The` and `the` are counted separately: 13 different words, and `was`
wins with 3); looping over `feedback` instead of `feedback.split()`, which counts **letters**.

`len(word_counts)` gives the number of different words because each word is one key. Alternative from Day 08:
`len(set(words))`.

## Exercise 6 (challenge): Letter bar chart

```python
phrase = "programming is fun"

letter_counts = {}
for ch in phrase:
    if ch != " ":
        letter_counts[ch] = letter_counts.get(ch, 0) + 1

# letter_counts
for letter in sorted(letter_counts):
    print(letter, "*" * letter_counts[letter])
```

Watch for students who sort `letter_counts.items()` (works, gives pairs) or who forget to skip spaces (an extra
line with a blank letter at the top). Extension: print the count after the bar, or use `phrase.replace(" ", "")`
before the loop instead of the `if`.
