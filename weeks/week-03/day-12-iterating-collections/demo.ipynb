## 1. Looping over a string

A string is a sequence of characters, so `for` visits them one at a time.

```python
for letter in "Python":
    print(letter)
```

### Vowel counter

Day 11's counter + `in` on a string (Day 06): `letter in "aeiou"` is `True` when the letter is one of those five.

```python
sentence = "Asha is learning Python"

vowels = 0
for letter in sentence.lower():
    if letter in "aeiou":
        vowels += 1
print("Vowels:", vowels)
```

🤔 **Predict:** without `.lower()`, is the answer the same?

```python
vowels = 0
for letter in sentence:
    if letter in "aeiou":
        vowels += 1
print("Vowels:", vowels)
```

The capital `A` in *Asha* is not in `"aeiou"`. Spaces are visited too; the `if` simply ignores them.

## 2. Looping over records: lists of dicts and `.items()`

Real data is usually a list of **records**, one dict per student, product or question.

```python
students = [
    {"name": "Asha", "mark": 81},
    {"name": "Omar", "mark": 47},
    {"name": "Mei", "mark": 66},
]

for student in students:
    print(student["name"], "scored", student["mark"])
    # print(student["name"], "scored", student["mark"])
```

On every pass `student` is a **whole dict**, so we use `student["name"]` inside the body.

### Looping over one dict: attendance report

Days present out of 20 school days.

```python
attendance = {"Asha": 19, "Bilal": 12, "Sara": 17, "Ken": 14}
attendance.items()
```

🤔 **Predict:** names or numbers?

```python
for name in attendance:
    print(name)
```

Looping over a dict gives its **keys**. To get key **and** value, use `.items()` (Day 08). Each item is a pair
like `("Asha", 19)`, unpacked into two names (Day 07).

```python
for name, days in attendance.items():
    percent = days / 20 * 100
    print(f"{name:<6} {days:>2}/20  {percent:5.1f}%")
```

`:<6` left-aligns the name in 6 characters (the mirror of `:>`). Use `.values()` when you only need the numbers:

```python
total_days = 0
for days in attendance.values():
    total_days += days
print("Average days present:", total_days / len(attendance))
```

## 3. `enumerate()`: position and item together

```python
tasks = ["buy milk", "email Ken", "finish homework"]

# for ind, val in enumerate(tasks, start=1):
#     print(ind)
#     print(val)

for number, task in enumerate(tasks, start=1):
    print(f"{number}. {task}")
```

🤔 **Predict:** what changes without `start=1`?

## 4. Building a new list

The **empty list + `.append`** pattern is an accumulator whose starting value is `[]`.

**Filter:** keep only some items.

```python
low_attendance = []                      # 1. empty list BEFORE the loop
for name, days in attendance.items():
    if days / 20 * 100 < 75:
        low_attendance.append(name)      # 2. append INSIDE the loop
print("Below 75%:", low_attendance)      # 3. use it AFTER the loop
```

**Transform:** a new value for every item.

```python
prices = [250, 1200, 80]
with_tax = []
for price in prices:
    with_tax.append(round(price * 1.1, 2))
print(with_tax)
```

## 5. Counting with a dict

Like a tally sheet: the first time we see a fruit we write it down with 1, every next time we add one.

```python
fruits = ["apple", "mango", "apple", "kiwi", "mango", "apple"]

counts = {}
for fruit in fruits:
    if fruit in counts:
        counts[fruit] += 1
    else:
        counts[fruit] = 1
print(counts)
```

Shorter, with `.get(key, default)` (Day 08): *"the old count, or 0 if it is new, plus one"*.

```python
freq = {}
for fruit in fruits:
    freq[fruit] = freq.get(fruit, 0) + 1
print(freq)
```

🤔 **Predict** the letter counts of `"letter"`:

```python
letter_counts = {}
for ch in "letter":
    letter_counts[ch] = letter_counts.get(ch, 0) + 1
print(letter_counts)
```

## 6. Finding the maximum by hand

`max(scores.values())` gives the top score, but not **who** got it. Keep the *best so far*, like a judge holding
the current winner's card and swapping it when someone beats it.

```python
scores = {"Diego": 72, "Fatima": 94, "Zara": 88}

best_name = ""
best_score = -1                     # lower than any real score
for name, score in scores.items():
    if score > best_score:
        best_score = score
        best_name = name
print(best_name, best_score)
```

| Pass | `name`, `score` | `score > best_score`? | best after the pass |
|------|-----------------|------------------------|---------------------|
| 1 | Diego, 72 | 72 > -1 ✔ | Diego 72 |
| 2 | Fatima, 94 | 94 > 72 ✔ | Fatima 94 |
| 3 | Zara, 88 | 88 > 94 ✘ | Fatima 94 |

🤔 **Predict:** two students tie. Who wins with `>`?

```python
scores = {"Diego": 94, "Fatima": 94, "Zara": 88}

best_name = ""
best_score = -1
for name, score in scores.items():
    if score > best_score:
        best_score = score
        best_name = name
print(best_name, best_score)
```

With `>` the **first** top scorer stays. With `>=` the last one would win.

## 7. Common mistakes (these cells fail **on purpose**)

```python
# Mistake 1: forgot .items(). Each key is a string like "Asha", not a (name, days) pair
for name, days in attendance:
    print(name, days)
```

```python
# Mistake 2: adding 1 to a count that does not exist yet
word_counts = {}
for word in ["hi", "hello", "hi"]:
    word_counts[word] += 1
```

```python
# Mistake 3: .append changes the list and gives back nothing
names = []
for name in ["Asha", "Bilal", "Sara"]:
    names = names.append(name)
```

**Fixes:** `attendance.items()` · `word_counts[word] = word_counts.get(word, 0) + 1` · just `names.append(name)`.

Mistake 3 fails on the **second** pass: after the first pass, `names` no longer holds a list.

A silent bug: starting `best_score = 0` when values can be negative (temperatures, bank balances).

```python
temperatures = {"Mon": -4, "Tue": -1, "Wed": -7}
warmest_day = ""
warmest = 0                 # bug: higher than every value!
for day, temp in temperatures.items():
    if temp > warmest:
        warmest = temp
        warmest_day = day
print("Warmest:", warmest_day, warmest)
```

Nothing is ever bigger than 0, so no day is chosen. Start lower than every possible value, or start with the first
item's value.

## 8. Real-world example: what do customers talk about?

A short product review. Which words appear most?

```python
review = "Great phone, great battery. The camera is good but the battery could be better. Great price."

clean = review.lower().replace(",", "").replace(".", "")
words = clean.split()
print(words)

word_counts = {}
for word in words:
    word_counts[word] = word_counts.get(word, 0) + 1
print(word_counts)
```

```python
top_word = ""
top_count = 0
for word, count in word_counts.items():
    if count > top_count:
        top_count = count
        top_word = word
print(f"Most common word: '{top_word}' ({top_count} times)")
```

Counting words is the simplest form of **text analysis**. In Week 11 we will compare this with what a language
model can do.
