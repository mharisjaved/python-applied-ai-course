## 1. Why `while`: a savings goal

Zara saves 650 a month. How many months until she has at least 5000?
We cannot use `range(...)` because we do not know the number of months yet: that is the question!

🤔 **Predict:** how many months?

````python
savings = 0
months = 0
while savings < 5000:
    savings += 650
    months += 1
print(f"{months} months, savings: {savings}")
````

`while condition:` means *"check the condition; if it is True, run the body; then check again"*.
After 7 months she has 4550 (still < 5000), so one more pass is needed.

## 2. Anatomy of a `while` loop

The same 1-to-5 count with `for` and with `while`:

````python
for count in range(1, 6):
    print(count)
````

````python
count = 1                 # 1. initialise
while count <= 5:         # 2. condition, checked BEFORE every pass
    print(count)
    count += 1            # 3. update: moves towards making the condition False
````

| Check | `count` | `count <= 5`? | Printed |
|-------|---------|----------------|---------|
| 1 | 1 | True | 1 |
| 2 | 2 | True | 2 |
| … | … | … | … |
| 5 | 5 | True | 5 |
| 6 | 6 | **False** → loop ends | – |

For a simple count, `for` is shorter. `while` shines when the end depends on what happens inside the loop.

🤔 **Predict:** how many times does the body run?

````python
count = 10
while count <= 5:
    print("inside the loop")
    count += 1
print("after the loop, count is", count)
````

Zero times: the condition is checked **before** the first pass, and it is already False.

## 3. Infinite loops (and how to stop them)

If nothing in the body makes the condition False, the loop never ends. This one forgot `count += 1`:

```python
count = 1
total = 0
while count <= 5:
    total += count
```

**Try it:** copy it into a new cell and run it. The cell shows `[*]` and never finishes.
- In a VS Code notebook: click **Interrupt** (■ at the top). You will see `KeyboardInterrupt`.
  If that does not help, click **Restart**.
- In a terminal (`uv run program.py`): press **Ctrl+C**.

🤔 **Predict (do not run!):** does this loop stop?

```python
n = 2
while n != 10:
    n += 3
```

`n` goes 2, 5, 8, 11, 14… and jumps over 10 forever. With `<` instead of `!=` it would stop:

````python
n = 2
while n < 10:
    n += 3
print(n)
````

## 4. Sentinel loops: "type q to quit"

A **sentinel** is a special value that means "stop". We ask once **before** the loop and again at the **end**
of every pass.

````python
shopping = []
item = input("Item (q to quit): ")          # first question, before the loop
while item != "q":
    shopping.append(item)
    item = input("Item (q to quit): ")      # ask again, at the end of the body
print(shopping)
print(len(shopping), "items")
````

- Delete the second `input` line and `item` never changes: infinite loop.
- If the user types `q` straight away, the body never runs and the list stays empty.

## 5. Validation loops with `.isdigit()`

On Day 03, `int("twenty")` crashed the program. `.isdigit()` lets us **check first**: it is `True` when every
character is a digit 0–9 (and the text is not empty).

🤔 **Predict** True or False for each, then run the cells one at a time.

````python
"42".isdigit()
````

````python
"4.5".isdigit()
````

````python
"-3".isdigit()
````

````python
"".isdigit()
````

````python
" 7".isdigit()
````

The dot, the minus sign and the space are not digits; an empty string has no digits at all.
For ages, quantities and PINs, that is exactly the check we want.

**The validation loop** has the same shape as the sentinel loop: ask, check, ask again.

````python
age_text = input("Your age: ")
while not age_text.isdigit():
    print("Please type a whole number, like 19.")
    age_text = input("Your age: ")

age = int(age_text)        # safe now: we know it is all digits
print(f"Next year you will be {age + 1}.")
````

## 6. PIN retry: limited attempts

The loop continues while the PIN is wrong **and** attempts are left. After the loop, an `if` works out *why* it
stopped.

````python
correct_pin = "2468"
attempts = 0
entered = ""

while entered != correct_pin and attempts < 3:
    entered = input("PIN: ")
    attempts += 1

if entered == correct_pin:
    print(f"Unlocked after {attempts} attempt(s).")
else:
    print("Card blocked.")
````

## 7. Common mistakes

````python
# Mistake 1: = assigns, == compares
count = 0
while count = 5:
    count += 1
````

````python
# Mistake 2: converting before validating
age_text = "twenty"    # pretend the user typed this
age = int(age_text)
````

Mistake 3 gives **no error**: forgetting the `()` after `isdigit`. Without brackets the method is not *called*,
and the method itself counts as "true".

````python
text = "abc"
if text.isdigit:
    print("Looks like a number?!")     # printed, wrongly
if text.isdigit():
    print("This line is not printed")
````

Mistake 4: no update, or no second `input()`, inside the loop → infinite loop (section 3).

## 8. `for` or `while`?

| Use `for` when… | Use `while` when… |
|-----------------|-------------------|
| you walk through a collection (list, string, dict) | you repeat **until** something changes |
| you know the number of repeats (`range`) | the number of repeats depends on the user or the data |
| *print every mark, total the prices* | *keep asking until valid, keep saving until the goal* |

## Recap

1. `while condition:` repeats while the condition is True; the condition is checked before every pass.
2. Three parts: **initialise, condition, update**. No update → infinite loop → Interrupt (notebook) or Ctrl+C.
3. Sentinel and validation loops: ask once before the loop, ask again at the end of the body.
4. Check text with `.isdigit()` **before** converting it with `int()`.
