## Exercise 1: Trace the while loops

| | Output | Trace | Common wrong answer |
|-|--------|-------|---------------------|
| a | `81` | 1 → 3 → 9 → 27 → 81 (81 < 50 is False) | `27` (stopping *before* passing 50) |
| b | `-1 3` | 5 → 3 → 1 → -1 | `1 2` or `0 3`: the check happens only at the top |
| c | `loop!!!!!!` | 4 letters + 6 `!` = 10 characters | 5 or 7 `!` |
| d | `done 0` | condition False at the first check | an infinite loop, or an error |

```python
x = 1
while x < 50:
    x = x * 3
print(x)

tickets = 5
sold = 0
while tickets > 0:
    tickets -= 2
    sold += 1
print(tickets, sold)

word = "loop"
while len(word) < 10:
    word = word + "!"
print(word)

count = 0
while count > 0:
    count -= 1
print("done", count)
```

(b) is the best discussion question: selling 2 tickets at a time from 5 overshoots to -1. In a real program the
condition should be `tickets >= 2`, or the last sale should be capped.

## Exercise 2: Growing plant

```python
height = 3
weeks = 0

while height < 30:
    height = height * 1.2
    weeks += 1

print(f"After {weeks} weeks the plant is {height:.1f} cm tall.")
print(weeks == 13)
```

Common mistakes: `height + 1.2` instead of `* 1.2` (grows by 1.2 cm per week: 23 weeks); `while height <= 30`
(same answer here, but say why `<` matches "at least 30"); printing inside the loop.

## Exercise 3: Fix the infinite loops

| | Problem | Fix |
|-|---------|-----|
| a | nothing changes `seconds` | `seconds -= 1` inside the body |
| b | the update goes the wrong way (1, 0, -1, … are all `<= 4`) | `level += 1` |
| c | 100 → 70 → 40 → 10 → -20: the balance jumps over 0 | `while balance > 0:` |

```python
level = 1
while level <= 4:
    print("Level", level)
    level += 1
```

```python
balance = 100
while balance > 0:
    balance -= 30
    print("Balance:", balance)
print("Card empty")
```

## Exercise 4: Score collector

Sentinel loop + three accumulators. The check for "no scores" must come **before** dividing, because
`total / count` with `count == 0` raises `ZeroDivisionError`.

```python
count = 0
total = 0
highest = -1

score_text = input("Score (or done): ")
while score_text != "done":
    score = int(score_text)
    count += 1
    total += score
    if score > highest:
        highest = score
    score_text = input("Score (or done): ")

if count == 0:
    print("No scores entered.")
else:
    print(f"{count} scores, total {total}, average {total / count:.2f}, highest {highest}")
```

Most common wrong answers:

- `int(input(...))` on the first line, which crashes when the user types `done`;
- adding the score *after* asking again, which adds `"done"` or skips the first score.

Fast students can skip invalid entries by only converting when `score_text.isdigit()` is true.

## Exercise 5: Ticket quantity

The condition reads: "keep asking while the text is **not** all digits, **or** the number is too small, **or** too
big". This is safe because `or` stops at the first part that is True: when the text is not digits, Python never
reaches `int(qty_text)`.

```python
qty_text = input("How many tickets (1-10)? ") # "four"
while not qty_text.isdigit() or int(qty_text) < 1 or int(qty_text) > 10:
    print("Please type a whole number from 1 to 10.")
    qty_text = input("How many tickets (1-10)? ")

quantity = int(qty_text)
print(f"{quantity} tickets cost {quantity * 12.50:.2f}")
```

```python
qty_text = "4"

print(not qty_text.isdigit()) # True
print(int(qty_text) < 1) # error
print(int(qty_text) > 10)
```

**Alternative** with a flag variable (`valid`). Longer, but some students find it easier to read because each test
is on its own line:

```python
valid = False
while not valid:
    qty_text = input("How many tickets (1-10)? ")
    if not qty_text.isdigit():
        print("Please type a whole number from 1 to 10.")
    elif int(qty_text) < 1 or int(qty_text) > 10:
        print("Please type a whole number from 1 to 10.")
    else:
        valid = True

quantity = int(qty_text)
print(f"{quantity} tickets cost {quantity * 12.50:.2f}")
```

Common mistakes:
- writing the range test first: `int(qty_text) < 1 or not qty_text.isdigit()` crashes on `three`, because `int()`
  runs before the digit check. A great example of *why the order matters*;
- `and` instead of `or` (accepts `12`, because `12` *is* digits);
- `1 < quantity < 10`, which rejects the valid answers 1 and 10.

## Exercise 6 (challenge): The 3n + 1 puzzle

```python
start = 27

n = start
steps = 0
highest = n
while n != 1:
    if n % 2 == 0:
        n = n // 2
    else:
        n = 3 * n + 1
    steps += 1
    if n > highest:
        highest = n

print(f"{start} reaches 1 after {steps} steps. Highest value: {highest}")
print(steps == 111, highest == 9232)
```

Here `!=` in the condition is fine: `n` changes by whole steps and is guaranteed (as far as anyone knows!) to hit 1
exactly. Ask students to compare this with Exercise 3c. Using `n / 2` instead of `n // 2` turns `n` into a float
(`13.0`); the loop still ends, but the printed values look odd. A good reminder of Day 03.
