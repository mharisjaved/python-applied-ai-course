## Exercise 2: Times table

`range(1, 13)` because the stop value is not included.

```python
number = 6

for i in range(1, 13):
    print(f"{number} x {i:>2} = {number * i:>3}")
```

praise the aligned one.

## Exercise 3: Fix the loop

| Mistake | Effect | Fix |
|---------|--------|-----|
| no `:` after the header | `SyntaxError` | `for price in prices:` |
| `print(Price)` | `NameError` (case matters) | `print(price)` |
| `total = 0` inside the loop | silent: reset every pass | move it before the loop |
| `total = total + price` not indented | silent: runs once, after the loop, with the last price | indent it into the body |

With only the two error fixes, the program prints `Total: 45`. Ask students *why* 45: it is a great tracing
question.

```python
prices = [120, 85, 300, 45]
total = 0
for price in prices:
    print(price)
    total = total + price
print("Total:", total)
print(total == 550)
```

## Exercise 4: Marks report

One loop, three accumulators. The count comes from `len(marks)` so the code adapts when the list changes.

```python
marks = [67, 48, 92, 55, 81, 39, 74, 50]

total = 0
passed = 0
distinctions = 0
for mark in marks:
    total += mark
    if mark >= 50:
        passed += 1
    if mark >= 80:
        distinctions += 1

average = total / len(marks)
print("Students:", len(marks))
print("Total:", total)
print("Average:", average)
print("Passed:", passed)
print("Distinctions:", distinctions)
```

```python
# Quick checks against the expected output
print(total == 506, passed == 6, distinctions == 2)
```

Most common wrong answers:
- `elif mark >= 80` after `if mark >= 50`: a distinction is also a pass, so the `elif` never runs and
  `Distinctions` is 0. Two separate `if`s are needed. (If students want `elif`, test the higher bar first and count
  distinctions as passes too.)
- `if mark > 50`: 50 is then not a pass, giving `Passed: 5`.
- printing inside the loop, so the report appears eight times.

## Exercise 5: Countdown and even numbers

```python
for n in range(10, 0, -1):
    print(n)
print("Happy New Year!")
```

```python
even_total = 0
for n in range(2, 101, 2):
    even_total += n
print("Sum of even numbers:", even_total)
print(even_total == 2550)
```

Common mistakes: `range(10, 1, -1)` stops at 2; `range(2, 100, 2)` leaves out 100 (sum 2450).

**Alternative** (teaches the same idea with an `if`): loop over `range(1, 101)` and add `n` only when `n % 2 == 0`
(`%` from Day 03). The step version is shorter and does half the work.

```python
even_total = 0
for n in range(1, 101):
    if n % 2 == 0:
        even_total += n
print(even_total)
```

## Exercise 6 (challenge): Savings plan

Two values change every month: what she saves (`saving`, grows by 50) and the running `total`.
The month number comes from `range(1, 13)`.

```python
saving = 200
total = 0
for month in range(1, 13):
    total += saving
    print(f"Month {month:>2}: saved {saving}, total {total}")
    saving += 50

print(f"After 12 months Mei has saved {total}.")
print(total == 5700)
```

The order inside the loop matters: add the saving to the total and print *before* increasing it, otherwise
month 1 shows 250.

**Alternative:** compute this month's saving from the month number: `saving = 200 + (month - 1) * 50`. That avoids
the second changing variable but needs a formula; both are fine.
