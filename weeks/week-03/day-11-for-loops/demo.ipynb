## 1. The problem: copy-paste repetition

Five students, five marks. We want to print each mark and whether it is a pass.

```python
marks = [72, 45, 88, 91, 60]

print(marks[0], marks[0] >= 50)
print(marks[1], marks[1] >= 50)
print(marks[2], marks[2] >= 50)
print(marks[3], marks[3] >= 50)
print(marks[4], marks[4] >= 50)
```

It works, but:
- 300 students would need 300 lines;
- if a student joins, we must change the code;
- a typo in one line (say `marks[3] >= 5`) is easy to miss.

We want to write the instruction **once** and let Python repeat it **for each** item.

## 2. The first `for` loop

Read it aloud: *"for each mark in marks, print the mark"*.

```python
marks = [72, 45, 88, 91, 60]

# names = ["Aisha", "Saim" ...]
# for name in names:

for x in marks:
    print(x)
    print("A number has been printed!.\n")
    
print("Done")
```

| Part | Meaning |
|------|---------|
| `for mark in marks:` | the header, ending with a colon |
| `mark` | the **loop variable**: a name tag that moves to the next item on every pass |
| indented lines | the **body**: runs once per item |
| `print("Done")` (not indented) | runs once, **after** the loop |

Trace of the first passes:

| Pass | `mark` | Printed |
|------|--------|---------|
| 1 | 72 | `72` |
| 2 | 45 | `45` |
| 3 | 88 | `88` |

🤔 **Predict:** how many lines will this print?

```python
marks = [72, 45, 88, 91, 60]

for mark in marks:
    if mark >= 50:
        print(mark, "pass")
```

45 fails the test, so the body runs 5 times but prints only 4 lines.

🤔 **Predict:** `print("Done")` is now indented. How many times does `Done` appear?

```python
for mark in marks:
    print(mark)
    print("Done")
```

**Indentation decides what repeats.** Everything indented under the header is the body.

The loop variable can have any name. Use the singular of the list name, so the code reads like English.

```python
guests = ["Asha", "Bilal", "Mei", "Diego"] # iteratable

# number = 87 # iterable?

for guest in guests:
    print(f"Welcome, {guest}!")
print(f"{len(guests)} guests have arrived.")
```

## 3. `range`: loops that count

Sometimes there is no list. We just want to repeat *n* times, or count 1, 2, 3…

```python
for i in range(5):
    print("Hello", i)
```

`range(5)` gives **five** numbers starting at **0**: 0, 1, 2, 3, 4.

To *see* the numbers a range produces, turn it into a list with `list(...)`:

```python
print(list(range(5)))
print(list(range(1, 6))) # string[start : stop : step]
print(list(range(0, 21, 5)))
print(list(range(10, 0, -1)))
```

| Form | Produces |
|------|----------|
| `range(stop)` | 0 … stop − 1 |
| `range(start, stop)` | start … stop − 1 (stop is **not** included, just like slicing) |
| `range(start, stop, step)` | start, start + step, … (negative step counts down) |

🤔 **Predict** each list before running:

```python
list(range(2, 11, 3))
```

```python
list(range(3))
```

11 is never reached in the first range. `range(3)` gives three numbers, but they are 0, 1, 2.

### A multiplication table

```python
number = 7
for i in range(1, 11):
    print(f"{number} x {i:>2} = {number * i:>3}")
```

`:>2` and `:>3` (Day 05) right-align the numbers so the columns line up.

### A countdown

```python
for n in range(5, 0, -1):
    print(n)
print("Lift off!")
```

## 4. The accumulator pattern

Like a till in a shop: the total starts at 0 **before** the first item, grows **while** scanning, and is read
**after** the last item.

```python
marks = [72, 45, 88, 91, 60]

total = 0 # 72, 117, 205, ...                 # 1. create the accumulator BEFORE the loop
for mark in marks:
    total += mark   # 2. update it INSIDE the loop
    print("Total: ", total)
    
print("Grand Total:", total)     # 3. use it AFTER the loop
```

| Pass | `mark` | `total` after the pass |
|------|--------|------------------------|
| start | – | 0 |
| 1 | 72 | 72 |
| 2 | 45 | 117 |
| 3 | 88 | 205 |
| 4 | 91 | 296 |
| 5 | 60 | 356 |

`total += mark` is the short form of `total = total + mark` (Day 03).

### Counting is the same pattern

```python
marks = [72, 45, 88, 91, 60]

count = 0
for mark in marks:
    count += 1
    print(count)
print("Number of marks:", count)
```

### Counting only some items: put the update inside an `if`

```python
marks = [72, 45, 88, 91, 60]

passed = 0
for mark in marks:
    if mark >= 50:
        passed += 1
print(passed, "of", len(marks), "passed")
```

### Total and count together give an average

```python
total = 0
count = 0
for mark in marks:
    total += mark
    count += 1

average = total / count    # after the loop!
print(f"Average: {average:.1f}")
```

🤔 **Predict:** `total = 0` has moved *inside* the loop. What is printed now?

```python
marks = [72, 45, 88, 91, 60]

total = 0
iteration = 0
for mark in marks:
    
    iteration += 1
    print(f"\n--------iteration 0{iteration}--------")
    print("Total before sum: ", total)
    
    print("total = total + mark")
    print(f"{total} = {total} + {mark}")
    total = total + mark
    
    print("Total after sum:", total)
    
print("Grand Total: ", total)
```

The total is reset on every pass, so only the last mark (60) survives. No error message: the program is just
wrong. **Accumulators go before the loop.**

`sum(marks)` gives a plain total, but it cannot count only the passes or add only some items. The pattern can.

## 5. Common mistakes (these cells fail **on purpose**)

Read the **last line** of each error.

```python
# Mistake 1: no colon at the end of the header
for mark in marks
    print(mark)
```

```python
# Mistake 2: the body is not indented
for mark in marks:
print(mark)
```

```python
# Mistake 3: input() always gives text, and range needs a whole number
times = "3"   # pretend this came from input()
for i in range(times):
    print("Hip hip hooray!")
```

**Fixes:** add `:` · indent the body (VS Code does it for you after the colon) · `range(int(times))`.

Two bugs that give **no error**:

```python
# Bug 4: we wanted 1 to 10, but stop is not included
for n in range(1, 10):
    print(n, end=" ")
```

`10` is missing: use `range(1, 11)`. (`end=" "` makes `print` finish with a space instead of a new line.)

Bug 5 is the accumulator inside the loop you saw in section 4. The most dangerous bugs are the silent ones:
**always check the result against a value you worked out by hand.**

## 6. Real-world example: one week of expenses

Bilal wrote down what he spent each day this week. His daily budget is 1000.

```python
expenses = [450, 1200, 80, 2300, 640, 150, 990]
daily_budget = 1000

total = 0
days_over = 0
for amount in expenses:
    total += amount
    if amount > daily_budget:
        days_over += 1

print(f"Spent this week: {total:,}")
print(f"Average per day: {total / len(expenses):.2f}")
print(f"Days over budget: {days_over}")
```

## Recap

1. A `for` loop runs its **indented body once for each item**. The loop variable holds the current item.
2. `range(start, stop, step)` counts from start up to **but not including** stop. `range(5)` is 0–4.
3. **Accumulator pattern:** create it before the loop, update it inside, use it after.
4. Silent bugs (off-by-one, reset inside the loop) give wrong answers without errors. Check by hand.
