# Day 15: How to Build a Project

**Today's question:** how do you go from a written brief to a working program without getting lost?

We will practise the process on a **tiny** program, a vocabulary flashcard drill. It is *not* the Quiz Game:
that one is yours to build.

## 1. The workflow

| Step | What it means |
|------|---------------|
| 1. **Read the brief** | Underline the *Must* requirements. Look at the example run. |
| 2. **Plan in comments** | Write the program's steps as `#` comments, in order. |
| 3. **Smallest working version** | The tiniest thing that runs and does something real. |
| 4. **Test** | Run it with normal, wrong and *odd* input. |
| 5. **Extend** | Add **one** feature → run → test → repeat. |

Skateboard → bicycle → car: each version works on its own.

## 2. Read the brief and plan

**Brief:** *Vocabulary flashcards. Show each Spanish word and ask for its English meaning. Say if the answer is
right (and give the meaning if not). At the end, show the score. Capital letters and extra spaces in the answer
should not matter.*

The plan is just comments. It runs, but does nothing yet, and that is fine.

```python
# PLAN: flashcard drill
# 1. Store the cards: a list of dicts, each with "word" and "meaning"
# 2. Set score to 0
# 3. For each card:
#    3a. Ask: What does <word> mean?
#    3b. Clean the answer (spaces, capitals)
#    3c. Compare with the meaning: say Correct, or give the meaning; add 1 to score if right
# 4. After the loop: print the score
#
# Smallest working version = steps 3a and 3c for ONE card
```

## 3. Smallest working version: one card

No list, no loop, no score yet. Just one question that really works.

```python
card = {"word": "gato", "meaning": "cat"}

answer = input(f"What does '{card['word']}' mean? ")
if answer == card["meaning"]:
    print("Correct!")
else:
    print(f"No, it means '{card['meaning']}'.")
```

It works! Before adding anything, **test** it.

## 4. Test like a real user

Real users type capitals and press the space bar. 🤔 **Predict:** what happens if the user types ` Cat`?

```python
card = {"word": "gato", "meaning": "cat"}

answer = input(f"What does '{card['word']}' mean? ")
if answer == card["meaning"]:
    print("Correct!")
else:
    print(f"No, it means '{card['meaning']}'.")
```

A bug! `" Cat"` is not equal to `"cat"`. Our first test (typing `cat`) could not find it; only **odd input** did.

Fix: clean the answer with `.strip()` and `.lower()` (Day 04) before comparing. Then run the same test again.

```python
card = {"word": "gato", "meaning": "cat"}

answer = input(f"What does '{card['word']}' mean? ").strip().lower()
if answer == card["meaning"]:
    print("Correct!")
else:
    print(f"No, it means '{card['meaning']}'.")
```

## 5. Extend: more cards, a loop and a score

One feature at a time: the single card becomes a list, the code moves into a `for` loop, and a score is added.
Everything inside the loop is the code we already tested.

```python
cards = [
    {"word": "gato", "meaning": "cat"},
    {"word": "perro", "meaning": "dog"},
]

score = 0
for number, card in enumerate(cards, start=1):
    answer = input(f"{number}/{len(cards)} What does '{card['word']}' mean? ").strip().lower()
    if answer == card["meaning"]:
        score += 1
        print("Correct!")
    else:
        print(f"No, it means '{card['meaning']}'.")

print(f"Score: {score}/{len(cards)}")
```

## 6. What next?

Possible next increments, each followed by a run and a test:
- more cards (only the list changes, because the loop uses `len(cards)`);
- `random.shuffle(cards)` for a new order each time;
- ask the wrong cards again at the end (build a list of wrong cards with `.append`, Day 12).
