## Exercise 2: First heatwave day

`enumerate(..., start=1)` (Day 12) gives the day number. The flag tells us after the loop whether `break` happened.

```python
temperatures = [31, 33, 34, 37, 36, 32, 38]
limit = 35

found = False
for day, temp in enumerate(temperatures, start=1):
    if temp > limit:
        print(f"First day above {limit}: day {day} ({temp}°C)")
        found = True
        break

if not found:
    print(f"No day above {limit}.")
```

```python
limit = 40

found = False
for day, temp in enumerate(temperatures, start=1):
    if temp > limit:
        print(f"First day above {limit}: day {day} ({temp}°C)")
        found = True
        break

if not found:
    print(f"No day above {limit}.")
```

Common mistakes: no `break` (prints days 4, 5 and 7); an `else: print("No day…")` **inside** the loop (prints it
for every cool day); using the index from 0 (`day 3`).

## Exercise 3: Step counter

```python
entries = ["8500", "", "10200", "n/a", "7300", "12000", "-"]

total = 0
valid_days = 0
skipped = 0
for entry in entries:
    if not entry.isdigit():
        skipped += 1
        continue
    total += int(entry)
    valid_days += 1

print("Valid days:", valid_days)
print("Total steps:", total)
print("Average per valid day:", total // valid_days)
print("Skipped entries:", skipped)
```

```python
print(total == 38000, valid_days == 4, skipped == 3)
```

`//` keeps the average a whole number (`9500`); `/` prints `9500.0`, which is also acceptable.
Common mistakes: dividing by `len(entries)` (7) instead of the valid days; `int(entry)` before the check, which
crashes on `""`; forgetting to count `skipped` *before* `continue` (anything after `continue` does not run).

**Alternative without `continue`:** `if entry.isdigit(): … else: skipped += 1`. Equally correct; `continue` keeps
the main work less indented, which matters more as the body grows.

## Exercise 4: Cinema seating plan

Outer loop over row letters (a string, Day 12), inner loop over seat numbers. Each row's labels are collected in a
list and joined.

```python
rows = "ABCD"
booked = ["A3", "B1", "B2", "D6"]

free = 0
for row in rows:
    labels = []
    for number in range(1, 7):
        seat = f"{row}{number}"
        if seat in booked:
            labels.append("XX")
        else:
            labels.append(seat)
            free += 1
    print(" ".join(labels))

print("Free seats:", free)
```

Common mistakes: `labels = []` *outside* the outer loop (each line repeats all previous seats); `range(1, 6)`
(only five seats); `free = 24 - len(booked)`, which works here but is not counting. Accept it and ask "what if a
booked seat were mistyped as `A9`?".

**Alternative:** build a string with `line += seat + " "`. Fine, but leaves a trailing space.

## Exercise 5: Guessing game with limited attempts

A `while` loop that runs while attempts are left, `continue` for bad input, `break` for a correct guess, and a
`won` flag to choose the final message.

```python
import random

random.seed(17)              # same secret while testing; delete this line to play for real
secret = random.randint(1, 50)
max_attempts = 6

attempts = 0
won = False
print(f"Guess the number (1-50). You have {max_attempts} attempts.")
while attempts < max_attempts:
    guess_text = input("Your guess: ")
    if not guess_text.isdigit():
        print("Whole numbers only, please.")
        continue
    guess = int(guess_text)
    attempts += 1
    if guess == secret:
        won = True
        break
    left = max_attempts - attempts
    if guess < secret:
        print(f"Too low! {left} attempts left.")
    else:
        print(f"Too high! {left} attempts left.")

if won:
    print(f"Correct! You got it in {attempts} attempts.")
else:
    print(f"Out of attempts! The number was {secret}.")
```

Most common wrong answers:
- `attempts += 1` *before* the digit check, so typing `forty` wastes an attempt;
- the secret created inside the loop (it changes every guess);
- checking `if guess == secret` *after* the loop instead of a flag: works, but is easy to get wrong when the last
  guess is invalid input;
- `while True` without any attempts check (unlimited game). Fine as a first step; then add the limit.

**Alternative:** `while True` with two exits (`break` on a correct guess, and `if attempts == max_attempts: break`).
Both versions are fine; the one above keeps "how long the game lasts" visible in the `while` line.

## Exercise 6 (challenge): Dice duel

```python
import random

random.seed(10)

asha_wins = 0
bilal_wins = 0
draws = 0
for round_number in range(1, 11):
    asha = random.randint(1, 6)
    bilal = random.randint(1, 6)
    if asha > bilal:
        winner = "Asha"
        asha_wins += 1
    elif bilal > asha:
        winner = "Bilal"
        bilal_wins += 1
    else:
        winner = "draw"
        draws += 1
    print(f"Round {round_number:>2}: Asha {asha}, Bilal {bilal} -> {winner}")

print(f"Asha {asha_wins} - Bilal {bilal_wins} ({draws} draw(s))")
if asha_wins > bilal_wins:
    print("Asha wins the duel!")
elif bilal_wins > asha_wins:
    print("Bilal wins the duel!")
else:
    print("The duel is a draw!")
```

If a student's rounds differ from the expected output, they almost always rolled both dice in a different order
or called `randint` an extra time (e.g. inside a `print`). This is a good way to show that a seed fixes the
**sequence** of random numbers, not each individual call.

Without the seed, over many runs both players win about equally often: the game is symmetric, so neither has an
advantage. Any streak is luck.
