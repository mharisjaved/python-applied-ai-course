## 1. `break`: leave the loop now

Find the **first** failing mark. Once we have found it, there is no reason to keep looking.

🤔 **Predict:** which lines are printed? Is 38 mentioned?

````python
marks = [72, 88, 45, 91, 38]

for mark in marks:
    if mark < 50:
        print("First fail found:", mark)
        break
    print(mark, "is fine")
    
print("Search finished")
````

`break` jumps straight to the first line **after** the loop. 91 and 38 are never visited.

### `while True` + `break`

Day 13's validation loop wrote `input()` twice. With `while True` the loop "runs forever", and the exit is a
`break` in the middle of the body, so `input()` is written once.

````python
while True:
    age_text = input("Your age: ")
    if age_text.isdigit():
        break
    print("Please type a whole number.")

age = int(age_text)
print("Age accepted:", age)
````

| Day 13 version | Day 14 version |
|----------------|----------------|
| ask before the loop, ask again at the end | ask once, at the top of the body |
| condition in the `while` line | condition in an `if` + `break` |

Both are correct. Choose the one that reads better for the problem.

## 2. `continue`: skip to the next pass

Like skipping a song in a playlist: `continue` jumps to the next item; `break` would stop the playlist.

🤔 **Predict:** what is the total, and how many "Skipping" lines are printed?

````python
readings = ["12", "", "15", "abc", "9"]

total = 0
for text in readings:
    if not text.isdigit():
        print(f"Skipping '{text}'")
        continue
    total += int(text)
print("Total:", total)
````

Everything after `continue` is skipped **for that pass only**. The bad entries never reach `int()`.

## 3. Nested loops: a loop inside a loop

Like a clock: for every hour, the minute hand goes all the way round. The **inner loop runs completely** for each
pass of the outer loop.

````python
for row in range(1, 4):   # row = 3 (first iteration)
    line = ""             # line = "3   6   9   12"
    for col in range(1, 6):   # col = 4
        line += f"{row * col:>4}"    # line = "" + f"3*4" -> "12"
    print(line)
````

`row` stays 1 while `col` goes 1, 2, 3, 4, 5. Only then does `row` become 2, and the inner loop starts again.
The inner body runs 3 × 5 = 15 times.

### Pairs: every match in a tournament

🤔 **Predict:** how many matches are printed?

````python
teams = ["Lions", "lion", "Tigers", "Bears"]

matches = 0
for home in teams:   # home = "Tigers"
    for away in teams:  # away = "Lions"
        if home != away:  #   Lion != lion -> True
            print(home, "vs", away)
            matches += 1
print(matches, "matches")
````

3 × 3 = 9 combinations, minus the 3 where a team would play itself.

## 4. The `random` module

A **module** is a toolbox of extra code that comes with Python. `import random` opens the toolbox; we use each tool
with `random.` in front. (On Day 26 we will import other modules and write our own.)

````python
import random

random.seed(6)
print(random.randint(1, 100))
````

Run the cell above a few times: the number changes. `randint(1, 6)` includes **both** 1 and 6
(unlike `range(1, 6)`, which stops at 5).

**`seed`** makes the sequence repeat. Run this cell several times: always the same three numbers.

````python
import random

random.seed(40)
print(random.randint(1, 100), random.randint(1, 100), random.randint(1, 100))
````

Computers make *pseudo*-random numbers from a starting value, the seed. Use a seed while **testing** and in demos;
remove it when people play for real.

````python
names = ["Asha", "Bilal", "Sara", "Omar", "Mei", "Bilal"]
random.seed(3)

print("Today's helper:", random.choice(names))

random.shuffle(names)      # changes the list in place
print("Presentation order:", names)

freq = {}

for name in names:
    freq[name] = freq.get(name, 0) + 1
    
print(freq)
````

Is a die fair? Roll it 600 times and count each face with a dict (Day 12):

````python
random.seed(5)
face_counts = {}
for roll in range(600):
    face = random.randint(1, 6)
    # print(face)
    face_counts[face] = face_counts.get(face, 0) + 1

for face in range(1, 7):
    print(face, face_counts[face])
    
print(face_counts)
````

Each face appears *about* 100 times, never exactly. That is what randomness looks like.

## 5. Live build: a number-guessing game

Built step by step: secret → loop with hints → `break` when correct → `continue` for bad input.

````python
random.seed(9)               # fixed secret for the demo; delete this line to play for real
secret = random.randint(1, 20)
# print(secret)
guesses = 0

print("I'm thinking of a number from 1 to 20.")
while True:
    guess_text = input("Your guess: ")
    if not guess_text.isdigit():
        print("Whole numbering of a number from 1 to 20 only, please.")
        continue                 # bad input does not count as a guess
    guess = int(guess_text)
    guesses += 1
    if guess < secret:
        print("Too low!")
    elif guess > secret:
        print("Too high!")
    else:
        print(f"Correct! You needed {guesses} guesses.")
        break
````

## 6. "Play again?": a loop around the game

The whole game goes inside an outer `while True`. The `break` for a correct guess leaves **only the inner loop**,
so we still reach the "Play again?" question.

````python
random.seed(7)

while True:                                  # outer loop: one pass = one game
    secret = random.randint(1, 20)
    guesses = 0
    print("New game! I'm thinking of a number from 1 to 20.")

    while True:                              # inner loop: one pass = one guess
        guess_text = input("Your guess: ")
        if not guess_text.isdigit():
            print("Whole numbers only, please.")
            continue
        guess = int(guess_text)
        guesses += 1
        if guess < secret:
            print("Too low!")
        elif guess > secret:
            print("Too high!")
        else:
            print(f"Correct! You needed {guesses} guesses.")
            break                            # leaves the inner loop only

    again = input("Play again? (y/n): ").lower()
    if again != "y":
        break                                # leaves the outer loop

print("Thanks for playing!")
````

Notice that `secret` and `guesses` are reset at the **start of each game**, inside the outer loop.

## 7. Common mistakes (these cells fail **on purpose**)

````python
# Mistake 1: the tool lives inside the module
roll = randint(1, 6)
````

````python
# Mistake 2: shuffle changes the list and gives back nothing
cards = ["ace", "king", "queen"]
cards = random.shuffle(cards)
print(cards[0])
````

````python
# Mistake 3: break only makes sense inside a loop
answer = "yes"
if answer == "yes":
    break
````

**Fixes:** `random.randint(1, 6)` · just `random.shuffle(cards)` on its own line · use `break` only in a loop.

Mistake 4 (do not run it): `continue` **before** the update in a `while` loop skips the update forever:

```python
n = 0
while n < 5:
    if n == 2:
        continue      # n stays 2 forever: infinite loop
    n += 1
```

## Recap

1. `break` leaves the loop immediately; `continue` skips to the next pass.
2. `while True` + `break` puts the exit test inside the body, so `input()` is written once.
3. Nested loops: the inner loop runs completely for every outer pass. `break` leaves only the innermost loop.
4. `random.randint(a, b)` includes both ends; `choice` picks one; `shuffle` mixes in place; `seed` repeats results.
